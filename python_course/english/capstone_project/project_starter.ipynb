# 💰 Project 1: Personal Expense Tracker

## Project Overview
Build a command-line application to track personal expenses, categorize spending, and generate financial reports.

## Features to Implement:
1. Add expenses with amount, category, and description
2. View all expenses
3. View expenses by category
4. Calculate total spending
5. Generate category summaries
6. Save/load data from JSON file
7. Handle errors gracefully

## Data Structure:
```python
expense = {
    'id': 1,
    'date': '2025-11-19',
    'amount': 50.00,
    'category': 'Food',
    'description': 'Grocery shopping'
}
```

---

## Step 1: Import Required Modules and Setup

````python
# Import required modules
import json
from datetime import datetime

# Global variables
expenses = []  # List to store all expenses
categories = {'Food', 'Transportation', 'Entertainment', 'Utilities', 'Shopping', 'Healthcare', 'Other'}
filename = 'expenses.json'

print("✅ Expense Tracker Setup Complete!")
````

## Step 2: File Operations Functions

````python
def save_expenses():
    """
    Save expenses to JSON file.
    Uses error handling to catch file writing errors.
    """
    # TODO: Implement save functionality
    # HINT: Use json.dump() with file opened in write mode
    # HINT: Use try-except to handle errors
    try:
        with open(filename, 'w') as file:
            json.dump(expenses, file, indent=4)
        print("💾 Expenses saved successfully!")
    except Exception as e:
        print(f"❌ Error saving expenses: {e}")


def load_expenses():
    """
    Load expenses from JSON file.
    If file doesn't exist, start with empty list.
    """
    global expenses
    # TODO: Implement load functionality
    # HINT: Use json.load() with file opened in read mode
    # HINT: Handle FileNotFoundError if file doesn't exist
    try:
        with open(filename, 'r') as file:
            expenses = json.load(file)
        print(f"✅ Loaded {len(expenses)} expenses from file.")
    except FileNotFoundError:
        print("📁 No existing expense file found. Starting fresh!")
        expenses = []
    except json.JSONDecodeError:
        print("⚠️ Error reading file. Starting with empty expense list.")
        expenses = []


# Test the functions
print("Testing file operations...")
load_expenses()
````

## Step 3: Core Expense Functions

````python
def add_expense():
    """
    Add a new expense with user input.
    Validates amount and category.
    """
    print("\n--- Add New Expense ---")
    
    # TODO: Get user input for amount
    # HINT: Use try-except to handle invalid number input
    try:
        amount = float(input("Enter amount: $"))
        if amount <= 0:
            print("❌ Amount must be positive!")
            return
    except ValueError:
        print("❌ Invalid amount! Please enter a number.")
        return
    
    # TODO: Get category from user
    # HINT: Show available categories
    # HINT: Validate category is in the allowed set
    print(f"\nCategories: {', '.join(sorted(categories))}")
    category = input("Enter category: ").strip().title()
    
    if category not in categories:
        print(f"⚠️ '{category}' not in categories. Adding to 'Other'.")
        category = 'Other'
    
    # TODO: Get description
    description = input("Enter description: ").strip()
    
    # TODO: Create expense dictionary
    # HINT: Use datetime.now() for current date
    # HINT: Use len(expenses) + 1 for ID
    expense = {
        'id': len(expenses) + 1,
        'date': datetime.now().strftime('%Y-%m-%d'),
        'amount': amount,
        'category': category,
        'description': description
    }
    
    # TODO: Add to expenses list
    expenses.append(expense)
    
    print(f"\n✅ Expense added successfully! (ID: {expense['id']})")
    save_expenses()  # Auto-save


def view_all_expenses():
    """
    Display all expenses in a formatted table.
    """
    # TODO: Check if expenses list is empty
    if not expenses:
        print("\n📭 No expenses recorded yet.")
        return
    
    print("\n" + "="*80)
    print("📊 ALL EXPENSES")
    print("="*80)
    print(f"{'ID':<5} {'Date':<12} {'Category':<15} {'Amount':<10} {'Description':<30}")
    print("-"*80)
    
    # TODO: Loop through expenses and display each one
    # HINT: Format amount as currency with 2 decimal places
    for expense in expenses:
        print(f"{expense['id']:<5} {expense['date']:<12} {expense['category']:<15} ${expense['amount']:<9.2f} {expense['description']:<30}")
    
    print("="*80)


def view_by_category():
    """
    Display expenses filtered by a specific category.
    """
    if not expenses:
        print("\n📭 No expenses recorded yet.")
        return
    
    # TODO: Get category from user
    print(f"\nCategories: {', '.join(sorted(categories))}")
    category = input("Enter category to view: ").strip().title()
    
    # TODO: Filter expenses by category
    # HINT: Use list comprehension or a for loop
    filtered = [exp for exp in expenses if exp['category'] == category]
    
    if not filtered:
        print(f"\n📭 No expenses found in '{category}' category.")
        return
    
    # TODO: Display filtered expenses
    print(f"\n📊 EXPENSES - {category.upper()} Category")
    print("-"*80)
    for expense in filtered:
        print(f"{expense['date']}: ${expense['amount']:.2f} - {expense['description']}")
    print("-"*80)


# Test adding an expense
print("\n🧪 Test: You can now test add_expense() function")
````

## Step 4: Calculation Functions

````python
def calculate_total():
    """
    Calculate and display total spending.
    """
    # TODO: Calculate sum of all expense amounts
    # HINT: Use a loop or sum() with list comprehension
    if not expenses:
        print("\n📭 No expenses to calculate.")
        return
    
    total = sum(expense['amount'] for expense in expenses)
    
    print("\n" + "="*40)
    print("💰 TOTAL SPENDING")
    print("="*40)
    print(f"Total Expenses: {len(expenses)}")
    print(f"Total Amount: ${total:.2f}")
    print(f"Average: ${total/len(expenses):.2f}")
    print("="*40)


def category_summary():
    """
    Calculate and display spending by category.
    """
    if not expenses:
        print("\n📭 No expenses to analyze.")
        return
    
    # TODO: Create a dictionary to store totals by category
    # HINT: Loop through expenses and add amounts to each category
    category_totals = {}
    
    for expense in expenses:
        cat = expense['category']
        if cat in category_totals:
            category_totals[cat] += expense['amount']
        else:
            category_totals[cat] = expense['amount']
    
    # TODO: Display category breakdown
    print("\n" + "="*50)
    print("📊 SPENDING BY CATEGORY")
    print("="*50)
    
    total = sum(category_totals.values())
    
    for cat in sorted(category_totals.keys()):
        amount = category_totals[cat]
        percentage = (amount / total) * 100
        print(f"{cat:<15} ${amount:>8.2f} ({percentage:>5.1f}%)")
    
    print("="*50)
    print(f"{'TOTAL':<15} ${total:>8.2f}")
    print("="*50)


# Test calculation functions
print("✅ Calculation functions defined!")
````

## Step 5: Main Menu System

````python
def display_menu():
    """
    Display the main menu options.
    """
    print("\n" + "="*40)
    print("💰 PERSONAL EXPENSE TRACKER")
    print("="*40)
    print("1. Add Expense")
    print("2. View All Expenses")
    print("3. View by Category")
    print("4. Calculate Total")
    print("5. Category Summary")
    print("6. Save & Exit")
    print("="*40)


def main():
    """
    Main program loop.
    """
    print("\n🎉 Welcome to Personal Expense Tracker!")
    load_expenses()
    
    # TODO: Create main menu loop
    # HINT: Use while True loop
    # HINT: Use conditionals to handle each menu choice
    # HINT: Use try-except to handle invalid menu choices
    
    while True:
        display_menu()
        
        try:
            choice = input("\nEnter your choice (1-6): ").strip()
            
            if choice == '1':
                add_expense()
            elif choice == '2':
                view_all_expenses()
            elif choice == '3':
                view_by_category()
            elif choice == '4':
                calculate_total()
            elif choice == '5':
                category_summary()
            elif choice == '6':
                save_expenses()
                print("\n👋 Thank you for using Expense Tracker! Goodbye!")
                break
            else:
                print("\n❌ Invalid choice! Please enter 1-6.")
        
        except KeyboardInterrupt:
            print("\n\n💾 Saving before exit...")
            save_expenses()
            print("👋 Goodbye!")
            break


# TODO: Uncomment the line below to run your program
# main()

print("\n✅ Expense Tracker Complete! Uncomment main() to run.")
````

# 📚 Project 2: Student Grade Manager

## Project Overview
Build a system to manage student records, track grades across multiple subjects, and generate academic reports.

## Features to Implement:
1. Add students with unique IDs
2. Record grades for multiple subjects
3. Calculate student averages
4. Calculate class averages by subject
5. Search for students
6. Save/load data from CSV file
7. Generate grade reports

## Data Structure:
```python
student = {
    'id': 'STU001',
    'name': 'Alice Johnson',
    'grades': {
        'Math': 85,
        'Science': 92,
        'English': 88
    }
}
```

---

## Step 1: Import Modules and Setup

````python
# Import required modules
import csv
import json

# Global variables
students = []  # List to store all student records
subjects = {'Math', 'Science', 'English', 'History', 'Art', 'PE'}
filename = 'students.json'  # Using JSON for easier nested data

print("✅ Student Grade Manager Setup Complete!")
````

## Step 2: File Operations

````python
def save_students():
    """
    Save student data to JSON file.
    """
    # TODO: Implement save functionality
    # HINT: Similar to expense tracker but saving students list
    try:
        with open(filename, 'w') as file:
            json.dump(students, file, indent=4)
        print("💾 Student data saved successfully!")
    except Exception as e:
        print(f"❌ Error saving data: {e}")


def load_students():
    """
    Load student data from JSON file.
    """
    global students
    # TODO: Implement load functionality
    try:
        with open(filename, 'r') as file:
            students = json.load(file)
        print(f"✅ Loaded {len(students)} students from file.")
    except FileNotFoundError:
        print("📁 No existing student file. Starting fresh!")
        students = []
    except json.JSONDecodeError:
        print("⚠️ Error reading file. Starting with empty list.")
        students = []


# Test file operations
load_students()
````

## Step 3: Student Management Functions

````python
def generate_student_id():
    """
    Generate a unique student ID.
    Format: STU001, STU002, etc.
    """
    # TODO: Generate ID based on number of students
    # HINT: Use f-string with zero padding
    num = len(students) + 1
    return f"STU{num:03d}"


def add_student():
    """
    Add a new student to the system.
    """
    print("\n--- Add New Student ---")
    
    # TODO: Get student name
    name = input("Enter student name: ").strip().title()
    
    if not name:
        print("❌ Name cannot be empty!")
        return
    
    # TODO: Generate student ID
    student_id = generate_student_id()
    
    # TODO: Create student dictionary
    student = {
        'id': student_id,
        'name': name,
        'grades': {}  # Empty grades dictionary
    }
    
    # TODO: Add to students list
    students.append(student)
    
    print(f"\n✅ Student added successfully!")
    print(f"   ID: {student_id}")
    print(f"   Name: {name}")
    save_students()


def find_student(student_id):
    """
    Find and return a student by ID.
    Returns None if not found.
    """
    # TODO: Search for student by ID
    # HINT: Loop through students list
    for student in students:
        if student['id'] == student_id:
            return student
    return None


def record_grade():
    """
    Record a grade for a student in a subject.
    """
    if not students:
        print("\n📭 No students in system. Add students first!")
        return
    
    print("\n--- Record Grade ---")
    
    # TODO: Get student ID
    student_id = input("Enter student ID: ").strip().upper()
    
    # TODO: Find student
    student = find_student(student_id)
    
    if not student:
        print(f"❌ Student {student_id} not found!")
        return
    
    # TODO: Get subject
    print(f"\nSubjects: {', '.join(sorted(subjects))}")
    subject = input("Enter subject: ").strip().title()
    
    if subject not in subjects:
        print(f"⚠️ Warning: '{subject}' not in standard subjects.")
        add_new = input("Add anyway? (y/n): ").lower()
        if add_new != 'y':
            return
    
    # TODO: Get grade with validation
    try:
        grade = float(input("Enter grade (0-100): "))
        if not 0 <= grade <= 100:
            print("❌ Grade must be between 0 and 100!")
            return
    except ValueError:
        print("❌ Invalid grade! Please enter a number.")
        return
    
    # TODO: Add grade to student's record
    student['grades'][subject] = grade
    
    print(f"\n✅ Grade recorded: {student['name']} - {subject}: {grade}")
    save_students()


def view_student_record():
    """
    Display a student's complete record.
    """
    if not students:
        print("\n📭 No students in system.")
        return
    
    # TODO: Get student ID and display their record
    student_id = input("\nEnter student ID: ").strip().upper()
    student = find_student(student_id)
    
    if not student:
        print(f"❌ Student {student_id} not found!")
        return
    
    # TODO: Display student info and all grades
    print("\n" + "="*50)
    print(f"📋 STUDENT RECORD")
    print("="*50)
    print(f"ID: {student['id']}")
    print(f"Name: {student['name']}")
    print("\nGrades:")
    
    if student['grades']:
        for subject, grade in sorted(student['grades'].items()):
            print(f"  {subject:<15} {grade:.1f}")
        
        # Calculate average
        avg = sum(student['grades'].values()) / len(student['grades'])
        print(f"\n  {'AVERAGE':<15} {avg:.2f}")
    else:
        print("  No grades recorded yet.")
    
    print("="*50)


print("✅ Student management functions defined!")
````

## Step 4: Statistics and Reports

````python
def calculate_student_average():
    """
    Calculate and display a student's grade average.
    """
    # TODO: Get student and calculate their average
    if not students:
        print("\n📭 No students in system.")
        return
    
    student_id = input("\nEnter student ID: ").strip().upper()
    student = find_student(student_id)
    
    if not student:
        print(f"❌ Student {student_id} not found!")
        return
    
    if not student['grades']:
        print(f"\n📭 No grades recorded for {student['name']}.")
        return
    
    # Calculate average
    avg = sum(student['grades'].values()) / len(student['grades'])
    
    print(f"\n📊 {student['name']}'s Average: {avg:.2f}")
    print(f"   Based on {len(student['grades'])} subjects")


def class_statistics():
    """
    Display statistics for the entire class.
    """
    # TODO: Calculate and display class statistics
    if not students:
        print("\n📭 No students in system.")
        return
    
    print("\n" + "="*50)
    print("📊 CLASS STATISTICS")
    print("="*50)
    
    # Total students
    print(f"Total Students: {len(students)}")
    
    # Students with grades
    students_with_grades = [s for s in students if s['grades']]
    print(f"Students with Grades: {len(students_with_grades)}")
    
    if not students_with_grades:
        print("\nNo grades recorded yet.")
        return
    
    # Calculate averages by subject
    print("\nAverage by Subject:")
    subject_totals = {}
    subject_counts = {}
    
    for student in students:
        for subject, grade in student['grades'].items():
            if subject not in subject_totals:
                subject_totals[subject] = 0
                subject_counts[subject] = 0
            subject_totals[subject] += grade
            subject_counts[subject] += 1
    
    for subject in sorted(subject_totals.keys()):
        avg = subject_totals[subject] / subject_counts[subject]
        print(f"  {subject:<15} {avg:.2f} (n={subject_counts[subject]})")
    
    print("="*50)


def search_student():
    """
    Search for students by name.
    """
    # TODO: Implement name search
    if not students:
        print("\n📭 No students in system.")
        return
    
    query = input("\nEnter name to search: ").strip().lower()
    
    # Find matching students
    matches = [s for s in students if query in s['name'].lower()]
    
    if not matches:
        print(f"\n❌ No students found matching '{query}'.")
        return
    
    print(f"\n🔍 Found {len(matches)} student(s):")
    print("-"*50)
    for student in matches:
        grade_count = len(student['grades'])
        print(f"{student['id']}: {student['name']} ({grade_count} grades)")
    print("-"*50)


print("✅ Statistics functions defined!")
````

## Step 5: Main Menu System

````python
def display_menu():
    """
    Display the main menu.
    """
    print("\n" + "="*40)
    print("📚 STUDENT GRADE MANAGER")
    print("="*40)
    print("1. Add Student")
    print("2. Record Grade")
    print("3. View Student Record")
    print("4. Calculate Student Average")
    print("5. Class Statistics")
    print("6. Search Student")
    print("7. Save & Exit")
    print("="*40)


def main():
    """
    Main program loop.
    """
    print("\n🎓 Welcome to Student Grade Manager!")
    load_students()
    
    # TODO: Create main menu loop
    # HINT: Similar structure to expense tracker
    while True:
        display_menu()
        
        try:
            choice = input("\nEnter your choice (1-7): ").strip()
            
            if choice == '1':
                add_student()
            elif choice == '2':
                record_grade()
            elif choice == '3':
                view_student_record()
            elif choice == '4':
                calculate_student_average()
            elif choice == '5':
                class_statistics()
            elif choice == '6':
                search_student()
            elif choice == '7':
                save_students()
                print("\n👋 Thank you for using Grade Manager! Goodbye!")
                break
            else:
                print("\n❌ Invalid choice! Please enter 1-7.")
        
        except KeyboardInterrupt:
            print("\n\n💾 Saving before exit...")
            save_students()
            print("👋 Goodbye!")
            break


# TODO: Uncomment to run your program
# main()

print("\n✅ Student Grade Manager Complete! Uncomment main() to run.")
````

# 📇 Project 3: Simple Contact Manager

## Project Overview
Create a contact management system to store, organize, search, and manage personal and professional contacts.

## Features to Implement:
1. Add contacts with multiple details
2. View all contacts
3. Search contacts by name, phone, or email
4. Edit contact information
5. Delete contacts
6. Categorize contacts (Family, Friends, Work)
7. Save/load from JSON
8. Export to CSV

## Data Structure:
```python
contact = {
    'id': 1,
    'name': 'John Doe',
    'phone': '555-1234',
    'email': 'john@example.com',
    'address': '123 Main St',
    'category': 'Friends'
}
```

---

## Step 1: Setup and Imports

````python
# Import required modules
import json
import csv

# Global variables
contacts = []  # List to store all contacts
categories = {'Family', 'Friends', 'Work', 'Other'}
json_filename = 'contacts.json'
csv_filename = 'contacts.csv'

print("✅ Contact Manager Setup Complete!")
````

## Step 2: File Operations

````python
def save_contacts():
    """
    Save contacts to JSON file.
    """
    # TODO: Implement save functionality
    try:
        with open(json_filename, 'w') as file:
            json.dump(contacts, file, indent=4)
        print("💾 Contacts saved successfully!")
    except Exception as e:
        print(f"❌ Error saving contacts: {e}")


def load_contacts():
    """
    Load contacts from JSON file.
    """
    global contacts
    # TODO: Implement load functionality
    try:
        with open(json_filename, 'r') as file:
            contacts = json.load(file)
        print(f"✅ Loaded {len(contacts)} contacts.")
    except FileNotFoundError:
        print("📁 No existing contact file. Starting fresh!")
        contacts = []
    except json.JSONDecodeError:
        print("⚠️ Error reading file. Starting with empty list.")
        contacts = []


def export_to_csv():
    """
    Export all contacts to CSV file.
    """
    # TODO: Implement CSV export
    if not contacts:
        print("\n📭 No contacts to export.")
        return
    
    try:
        with open(csv_filename, 'w', newline='') as file:
            # Define CSV fields
            fieldnames = ['id', 'name', 'phone', 'email', 'address', 'category']
            writer = csv.DictWriter(file, fieldnames=fieldnames)
            
            # Write header
            writer.writeheader()
            
            # Write all contacts
            writer.writerows(contacts)
        
        print(f"\n✅ Exported {len(contacts)} contacts to {csv_filename}")
    except Exception as e:
        print(f"❌ Error exporting to CSV: {e}")


# Test file operations
load_contacts()
````

## Step 3: Contact CRUD Operations

````python
def add_contact():
    """
    Add a new contact with user input.
    """
    print("\n--- Add New Contact ---")
    
    # TODO: Get contact information from user
    name = input("Enter name: ").strip().title()
    if not name:
        print("❌ Name cannot be empty!")
        return
    
    phone = input("Enter phone number: ").strip()
    email = input("Enter email: ").strip().lower()
    address = input("Enter address: ").strip()
    
    # Get category
    print(f"\nCategories: {', '.join(sorted(categories))}")
    category = input("Enter category: ").strip().title()
    
    if category not in categories:
        print(f"⚠️ '{category}' not in categories. Using 'Other'.")
        category = 'Other'
    
    # TODO: Create contact dictionary
    contact = {
        'id': len(contacts) + 1,
        'name': name,
        'phone': phone,
        'email': email,
        'address': address,
        'category': category
    }
    
    # TODO: Add to contacts list
    contacts.append(contact)
    
    print(f"\n✅ Contact added successfully! (ID: {contact['id']})")
    save_contacts()


def view_all_contacts():
    """
    Display all contacts in a formatted list.
    """
    # TODO: Display all contacts
    if not contacts:
        print("\n📭 No contacts saved yet.")
        return
    
    print("\n" + "="*80)
    print("📇 ALL CONTACTS")
    print("="*80)
    
    for contact in contacts:
        print(f"\nID: {contact['id']}")
        print(f"Name: {contact['name']}")
        print(f"Phone: {contact['phone']}")
        print(f"Email: {contact['email']}")
        print(f"Address: {contact['address']}")
        print(f"Category: {contact['category']}")
        print("-"*40)
    
    print(f"\nTotal Contacts: {len(contacts)}")
    print("="*80)


def find_contact(contact_id):
    """
    Find and return a contact by ID.
    """
    # TODO: Search for contact by ID
    for contact in contacts:
        if contact['id'] == contact_id:
            return contact
    return None


def edit_contact():
    """
    Edit an existing contact's information.
    """
    if not contacts:
        print("\n📭 No contacts to edit.")
        return
    
    print("\n--- Edit Contact ---")
    
    # TODO: Get contact ID
    try:
        contact_id = int(input("Enter contact ID to edit: "))
    except ValueError:
        print("❌ Invalid ID!")
        return
    
    # TODO: Find contact
    contact = find_contact(contact_id)
    
    if not contact:
        print(f"❌ Contact ID {contact_id} not found!")
        return
    
    # Display current info
    print(f"\nCurrent information for {contact['name']}:")
    print(f"Phone: {contact['phone']}")
    print(f"Email: {contact['email']}")
    print(f"Address: {contact['address']}")
    print(f"Category: {contact['category']}")
    
    # TODO: Get new information (press Enter to keep current)
    print("\nEnter new information (press Enter to keep current):")
    
    new_phone = input(f"Phone [{contact['phone']}]: ").strip()
    if new_phone:
        contact['phone'] = new_phone
    
    new_email = input(f"Email [{contact['email']}]: ").strip().lower()
    if new_email:
        contact['email'] = new_email
    
    new_address = input(f"Address [{contact['address']}]: ").strip()
    if new_address:
        contact['address'] = new_address
    
    print(f"\nCategories: {', '.join(sorted(categories))}")
    new_category = input(f"Category [{contact['category']}]: ").strip().title()
    if new_category and new_category in categories:
        contact['category'] = new_category
    
    print("\n✅ Contact updated successfully!")
    save_contacts()


def delete_contact():
    """
    Delete a contact from the list.
    """
    if not contacts:
        print("\n📭 No contacts to delete.")
        return
    
    print("\n--- Delete Contact ---")
    
    # TODO: Get contact ID and confirm deletion
    try:
        contact_id = int(input("Enter contact ID to delete: "))
    except ValueError:
        print("❌ Invalid ID!")
        return
    
    contact = find_contact(contact_id)
    
    if not contact:
        print(f"❌ Contact ID {contact_id} not found!")
        return
    
    # Confirm deletion
    print(f"\n⚠️ Are you sure you want to delete '{contact['name']}'?")
    confirm = input("Type 'yes' to confirm: ").lower()
    
    if confirm == 'yes':
        contacts.remove(contact)
        print(f"\n✅ Contact '{contact['name']}' deleted successfully.")
        save_contacts()
    else:
        print("\n❌ Deletion cancelled.")


print("✅ CRUD functions defined!")
````

## Step 4: Search and Filter Functions

````python
def search_contacts():
    """
    Search contacts by name, phone, or email.
    """
    if not contacts:
        print("\n📭 No contacts to search.")
        return
    
    # TODO: Get search query
    query = input("\nEnter name, phone, or email to search: ").strip().lower()
    
    # TODO: Search in all fields
    matches = []
    for contact in contacts:
        if (query in contact['name'].lower() or 
            query in contact['phone'].lower() or 
            query in contact['email'].lower()):
            matches.append(contact)
    
    # TODO: Display results
    if not matches:
        print(f"\n❌ No contacts found matching '{query}'.")
        return
    
    print(f"\n🔍 Found {len(matches)} contact(s):")
    print("="*60)
    
    for contact in matches:
        print(f"\nID: {contact['id']}")
        print(f"Name: {contact['name']}")
        print(f"Phone: {contact['phone']}")
        print(f"Email: {contact['email']}")
        print(f"Category: {contact['category']}")
        print("-"*40)


def view_by_category():
    """
    Display contacts filtered by category.
    """
    if not contacts:
        print("\n📭 No contacts saved yet.")
        return
    
    # TODO: Get category and filter contacts
    print(f"\nCategories: {', '.join(sorted(categories))}")
    category = input("Enter category to view: ").strip().title()
    
    filtered = [c for c in contacts if c['category'] == category]
    
    if not filtered:
        print(f"\n📭 No contacts in '{category}' category.")
        return
    
    # TODO: Display filtered contacts
    print(f"\n📇 {category.upper()} CONTACTS")
    print("="*60)
    
    for contact in filtered:
        print(f"\n{contact['name']}")
        print(f"  Phone: {contact['phone']}")
        print(f"  Email: {contact['email']}")
    
    print(f"\nTotal: {len(filtered)} contacts")
    print("="*60)


print("✅ Search functions defined!")
````

## Step 5: Main Menu System

````python
def display_menu():
    """
    Display the main menu.
    """
    print("\n" + "="*40)
    print("📇 CONTACT MANAGER")
    print("="*40)
    print("1. Add Contact")
    print("2. View All Contacts")
    print("3. Search Contact")
    print("4. Edit Contact")
    print("5. Delete Contact")
    print("6. View by Category")
    print("7. Export to CSV")
    print("8. Save & Exit")
    print("="*40)


def main():
    """
    Main program loop.
    """
    print("\n📇 Welcome to Contact Manager!")
    load_contacts()
    
    # TODO: Create main menu loop
    while True:
        display_menu()
        
        try:
            choice = input("\nEnter your choice (1-8): ").strip()
            
            if choice == '1':
                add_contact()
            elif choice == '2':
                view_all_contacts()
            elif choice == '3':
                search_contacts()
            elif choice == '4':
                edit_contact()
            elif choice == '5':
                delete_contact()
            elif choice == '6':
                view_by_category()
            elif choice == '7':
                export_to_csv()
            elif choice == '8':
                save_contacts()
                print("\n👋 Thank you for using Contact Manager! Goodbye!")
                break
            else:
                print("\n❌ Invalid choice! Please enter 1-8.")
        
        except KeyboardInterrupt:
            print("\n\n💾 Saving before exit...")
            save_contacts()
            print("👋 Goodbye!")
            break


# TODO: Uncomment to run your program
# main()

print("\n✅ Contact Manager Complete! Uncomment main() to run.")
````
