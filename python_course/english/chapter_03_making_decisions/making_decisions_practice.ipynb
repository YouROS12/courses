## Part 2: Comparison Operators - Making Comparisons ⚖️

Comparison operators let you compare values and return True or False.

### Equal To (==)

```python
# Check if two values are equal
print(5 == 5)        # True
print(5 == 10)       # False
print("hi" == "hi")  # True
print("hi" == "Hi")  # False (case matters!)
```

```python
# Using variables
age = 18
voting_age = 18

result = age == voting_age
print(f"Is age equal to voting age? {result}")
```

### Not Equal To (!=)

```python
# Check if two values are NOT equal
print(5 != 10)       # True
print(5 != 5)        # False
print("cat" != "dog") # True
```

```python
# Practical example
password = "secret123"
user_input = "Secret123"

print(f"Passwords match? {password == user_input}")  # False
print(f"Passwords different? {password != user_input}")  # True
```

### Greater Than (>) and Less Than (<)

```python
# Greater than
print(10 > 5)        # True
print(5 > 10)        # False
print(5 > 5)         # False (not greater, they're equal)
```

```python
# Less than
print(5 < 10)        # True
print(10 < 5)        # False
print(5 < 5)         # False
```

```python
# Using variables
temperature = 35
hot_threshold = 30

is_hot = temperature > hot_threshold
print(f"Temperature: {temperature}°C")
print(f"Is it hot? {is_hot}")
```

### Greater Than or Equal To (>=) and Less Than or Equal To (<=)

```python
# Greater than or equal to
print(10 >= 5)       # True
print(10 >= 10)      # True (equal counts!)
print(5 >= 10)       # False
```

```python
# Less than or equal to
print(5 <= 10)       # True
print(10 <= 10)      # True
print(15 <= 10)      # False
```

```python
# Practical example - age verification
age = 18
minimum_age = 18

can_vote = age >= minimum_age
print(f"Age: {age}")
print(f"Can vote? {can_vote}")
```

## Part 3: If Statements - Making Decisions 🛤️

If statements let your program make decisions and execute code only when a condition is True.

```python
# Basic if statement
age = 20

if age >= 18:
    print("You are an adult!")  # This runs only if condition is True
    print("You can vote!")
```

```python
# When condition is False, the code inside is skipped
age = 15

if age >= 18:
    print("You are an adult!")  # This won't run

print("This always runs!")  # This runs regardless
```

```python
# Multiple if statements
temperature = 35

if temperature > 30:
    print("It's hot outside!")

if temperature > 40:
    print("It's extremely hot!")

if temperature < 0:
    print("It's freezing!")
```

```python
# If statement with user input
score = int(input("Enter your test score: "))

if score >= 90:
    print("Excellent! You got an A!")
```

## Part 6: Logical Operators - Combining Conditions 🔗

Logical operators let you combine multiple conditions.

### AND Operator - Both conditions must be True

```python
# Basic AND
print(True and True)      # True
print(True and False)     # False
print(False and True)     # False
print(False and False)    # False
```

```python
# Practical example
age = 25
has_license = True

if age >= 18 and has_license:
    print("You can drive!")
else:
    print("You cannot drive.")
```

```python
# Range checking
temperature = 22

if temperature >= 20 and temperature <= 25:
    print("Perfect temperature!")
else:
    print("Not ideal temperature")
```

```python
# Multiple conditions with AND
age = 20
gpa = 3.5
attendance = 90

if age >= 18 and gpa >= 3.0 and attendance >= 80:
    print("Eligible for scholarship!")
else:
    print("Not eligible for scholarship.")
```

### OR Operator - At least one condition must be True

```python
# Basic OR
print(True or True)       # True
print(True or False)      # True
print(False or True)      # True
print(False or False)     # False
```

```python
# Practical example
day = "Saturday"

if day == "Saturday" or day == "Sunday":
    print("It's the weekend! 🎉")
else:
    print("It's a weekday.")
```

```python
# Emergency detection
temperature = 45
smoke_detected = False

if temperature > 40 or smoke_detected:
    print("⚠️ ALERT! Possible fire hazard!")
else:
    print("✓ All systems normal")
```

### NOT Operator - Reverses the condition

```python
# Basic NOT
print(not True)       # False
print(not False)      # True
```

```python
# Practical example
is_raining = False

if not is_raining:
    print("Great day for a picnic!")
else:
    print("Better stay inside.")
```

```python
# Access control
is_banned = False
age = 20

if not is_banned and age >= 18:
    print("Access granted")
else:
    print("Access denied")
```

### Combining Multiple Logical Operators

```python
# Complex conditions with parentheses
age = 25
income = 50000
credit_score = 720

# Loan eligibility checker
if (age >= 21 and age <= 65) and (income >= 30000 or credit_score >= 700):
    print("✓ Loan approved!")
else:
    print("✗ Loan denied.")
```

```python
# Store hours checker
day = "Saturday"
hour = 14

if (day == "Saturday" or day == "Sunday") and (hour >= 9 and hour < 18):
    print("Store is OPEN")
elif day not in ["Saturday", "Sunday"] and (hour >= 8 and hour < 20):
    print("Store is OPEN")
else:
    print("Store is CLOSED")
```

### Example 1: Smart Thermostat

```python
# Smart home temperature controller
current_temp = float(input("Enter current temperature (°C): "))
desired_temp = float(input("Enter desired temperature (°C): "))

print("\n--- SMART THERMOSTAT ---")

if current_temp < desired_temp - 2:
    print("🔥 Turning heating ON (HIGH)")
elif current_temp < desired_temp:
    print("🔥 Turning heating ON (LOW)")
elif current_temp > desired_temp + 2:
    print("❄️ Turning cooling ON (HIGH)")
elif current_temp > desired_temp:
    print("❄️ Turning cooling ON (LOW)")
else:
    print("✓ Temperature is perfect! System idle.")
```

### Example 2: Discount Calculator

```python
# Shopping discount calculator
total_purchase = float(input("Enter total purchase amount: $"))
is_member = input("Are you a member? (yes/no): ").lower() == "yes"
has_coupon = input("Do you have a coupon? (yes/no): ").lower() == "yes"

discount = 0

# Determine discount based on conditions
if total_purchase >= 100 and is_member:
    discount = 20  # 20% for members spending $100+
elif total_purchase >= 100:
    discount = 15  # 15% for non-members spending $100+
elif is_member:
    discount = 10  # 10% for members
elif total_purchase >= 50:
    discount = 5   # 5% for purchases over $50

# Extra 5% if they have a coupon
if has_coupon:
    discount += 5

# Calculate final price
discount_amount = total_purchase * (discount / 100)
final_price = total_purchase - discount_amount

# Display results
print("\n" + "=" * 40)
print("          PURCHASE SUMMARY")
print("=" * 40)
print(f"Original Price:    ${total_purchase:.2f}")
print(f"Discount:          {discount}% (${discount_amount:.2f})")
print(f"Final Price:       ${final_price:.2f}")
print("=" * 40)

if discount > 0:
    print(f"🎉 You saved ${discount_amount:.2f}!")
else:
    print("💡 Tip: Become a member or spend more to get discounts!")
```

### Example 3: BMI Calculator with Health Advice

```python
# BMI Calculator with health categories
weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))

# Calculate BMI
bmi = weight / (height ** 2)

print(f"\nYour BMI is: {bmi:.1f}")

# Classify BMI and give advice
if bmi < 18.5:
    category = "Underweight"
    advice = "Consider eating more nutritious foods and consulting a nutritionist."
elif bmi >= 18.5 and bmi < 25:
    category = "Normal weight"
    advice = "Great! Maintain your healthy lifestyle."
elif bmi >= 25 and bmi < 30:
    category = "Overweight"
    advice = "Consider a balanced diet and regular exercise."
else:
    category = "Obese"
    advice = "Consult a healthcare provider for personalized advice."

print(f"Category: {category}")
print(f"Advice: {advice}")
```

### Example 5: Exam Grade Analyzer

```python
# Comprehensive grade analyzer
score = int(input("Enter your exam score (0-100): "))

# Validate input
if score < 0 or score > 100:
    print("❌ Invalid score! Must be between 0 and 100.")
else:
    # Determine grade
    if score >= 90:
        grade = "A"
        message = "Outstanding! Excellent work!"
        emoji = "🌟"
    elif score >= 80:
        grade = "B"
        message = "Great job! Keep it up!"
        emoji = "👍"
    elif score >= 70:
        grade = "C"
        message = "Good work. Room for improvement."
        emoji = "📚"
    elif score >= 60:
        grade = "D"
        message = "You passed, but need to study more."
        emoji = "⚠️"
    else:
        grade = "F"
        message = "Failed. Don't give up - study and try again!"
        emoji = "💪"
    
    # Display results
    print("\n" + "=" * 50)
    print(f"  EXAM RESULTS {emoji}")
    print("=" * 50)
    print(f"  Score: {score}/100")
    print(f"  Grade: {grade}")
    print(f"  {message}")
    print("=" * 50)
    
    # Additional feedback
    if score >= 90:
        print("\n🎓 You're in the top tier!")
    elif score >= 70:
        print(f"\n📈 You need {90 - score} more points for an A")
    elif score >= 60:
        print(f"\n⚡ You need {70 - score} more points to reach a C")
    else:
        print(f"\n💡 You need {60 - score} more points to pass")
```

### Example 6: Shipping Cost Calculator

```python
# E-commerce shipping calculator
weight = float(input("Enter package weight (kg): "))
distance = float(input("Enter distance (km): "))
is_express = input("Express shipping? (yes/no): ").lower() == "yes"

# Base cost calculation
if weight <= 1:
    base_cost = 5
elif weight <= 5:
    base_cost = 10
elif weight <= 10:
    base_cost = 15
else:
    base_cost = 15 + (weight - 10) * 2  # $2 per kg over 10kg

# Distance multiplier
if distance <= 50:
    distance_multiplier = 1.0
elif distance <= 200:
    distance_multiplier = 1.5
else:
    distance_multiplier = 2.0

# Calculate total
shipping_cost = base_cost * distance_multiplier

# Express shipping adds 50%
if is_express:
    shipping_cost *= 1.5
    delivery_time = "1-2 days"
else:
    delivery_time = "5-7 days"

# Free shipping for orders over $100
if shipping_cost > 100:
    print("\n🎉 FREE SHIPPING! (Orders over $100)")
    shipping_cost = 0

# Display results
print("\n" + "=" * 40)
print("       SHIPPING INFORMATION")
print("=" * 40)
print(f"Weight:          {weight} kg")
print(f"Distance:        {distance} km")
print(f"Express:         {'Yes' if is_express else 'No'}")
print(f"Delivery Time:   {delivery_time}")
print(f"Shipping Cost:   ${shipping_cost:.2f}")
print("=" * 40)
```
