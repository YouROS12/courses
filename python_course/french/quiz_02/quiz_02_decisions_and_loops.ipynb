### Part E: Debugging - Solutions

**Debug 1:** Missing colon after if condition
```python
age = 15
if age >= 18:  # Added colon
    print("Adult")
```

**Debug 2:** Missing colon and wrong indentation
```python
for i in range(5):  # Added colon
    print(i)  # Added indentation
```

**Debug 3:** Infinite loop - count never increases
```python
count = 0
while count < 5:
    print(count)
    count += 1  # Added increment
```

**Debug 4:** Missing colon after second elif
```python
score = 75
if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:  # Added colon
    print("C")
```

**Debug 5:** Using = (assignment) instead of == (comparison)
```python
x = 10
if x == 10:  # Changed = to ==
    print("x is 10")
```

**Debug 6:** Inconsistent indentation
```python

    print(i)  # Added indentation
    print("Done")  # Already indented correctly
# OR if "Done" should print after loop:
# for i in range(3):
#     print(i)
# print("Done")  # Remove indentation
```
