## Partie 2 : Boucles While - Répéter Tant Que Vrai ⏳

Une **boucle while** répète tant qu'une condition est True.

### Structure Basique de la Boucle While

```python
while condition:
    # code à répéter
```

````python
# Count from 1 to 5
count = 1

while count <= 5:
    print(count)
    count = count + 1  # IMPORTANT: Update the counter!

print("Done counting!")
````

````python
# Countdown example
countdown = 10

while countdown > 0:
    print(countdown)
    countdown -= 1  # Same as: countdown = countdown - 1

print("Liftoff! 🚀")
````

````python
# Sum numbers from 1 to 10
number = 1
total = 0

while number <= 10:
    total += number  # Add number to total
    number += 1

print(f"Sum of 1 to 10 = {total}")
````

### Boucles While avec Entrée Utilisateur

````python
# Keep asking until user enters 'quit'
user_input = ""

while user_input != "quit":
    user_input = input("Enter 'quit' to exit: ")
    print(f"You entered: {user_input}")

print("Goodbye!")
````

````python
# Password validator
password = ""
correct_password = "python123"

while password != correct_password:
    password = input("Enter password: ")
    if password != correct_password:
        print("Incorrect! Try again.")

print("Access granted! ✅")
````

## Partie 3 : Boucles For - Itérer à Travers des Séquences 🔁

Une **boucle for** itère à travers chaque élément d'une séquence.

### Boucler à Travers des Chaînes

````python
# Print each character in a string
word = "Python"

for letter in word:
    print(letter)
````

````python
# Count vowels in a word
word = "Programming"
vowels = "aeiouAEIOU"
count = 0

for letter in word:
    if letter in vowels:
        count += 1

print(f"'{word}' has {count} vowels")
````

### Boucler à Travers des Listes

````python
# Iterate through a list of numbers
numbers = [10, 20, 30, 40, 50]

for num in numbers:
    print(num)
````

````python
# Calculate sum of a list
numbers = [5, 10, 15, 20, 25]
total = 0

for num in numbers:
    total += num

print(f"Sum: {total}")
print(f"Average: {total / len(numbers)}")
````

````python
# Find maximum in a list
numbers = [45, 12, 78, 34, 90, 23]
max_num = numbers[0]  # Start with first number

for num in numbers:
    if num > max_num:
        max_num = num

print(f"Maximum number: {max_num}")
````

## Partie 4 : La Fonction range() 📊

La fonction **range()** génère une séquence de nombres - parfaite pour les boucles !

### range(stop) - Un Paramètre

````python
# range(5) generates: 0, 1, 2, 3, 4 (stops BEFORE 5)
for i in range(5):
    print(i)
````

````python
# Print "Hello" 10 times
for i in range(10):
    print(f"Hello #{i+1}")
````

### range(start, stop) - Deux Paramètres

````python
# range(3, 8) generates: 3, 4, 5, 6, 7
for i in range(3, 8):
    print(i)
````

````python
# Print numbers 1 to 10 (inclusive)
for i in range(1, 11):
    print(i)
````

### range(start, stop, step) - Trois Paramètres

````python
# Count by 2s: range(0, 10, 2) generates: 0, 2, 4, 6, 8
for i in range(0, 10, 2):
    print(i)
````

````python
# Odd numbers from 1 to 20
for i in range(1, 21, 2):
    print(i, end=" ")
print()  # New line at end
````

````python
# Count by 10s
for i in range(0, 101, 10):
    print(i, end=" ")
print()
````

````python
# Countdown with negative step
for i in range(10, 0, -1):
    print(i, end=" ")
print("Liftoff! 🚀")
````

### Modèles Utiles de range()

````python
# Sum of 1 to 100
total = 0
for i in range(1, 101):
    total += i
print(f"Sum of 1 to 100 = {total}")
````

````python
# Multiplication table for 7
for i in range(1, 11):
    print(f"7 x {i} = {7 * i}")
````

````python
# Even numbers backwards
for i in range(20, 0, -2):
    print(i, end=" ")
print()
````

## Partie 5 : Instruction Break - Sortir Tôt 🛑

L'instruction **break** quitte immédiatement la boucle.

````python
# Stop when we find number 7
for i in range(1, 11):
    if i == 7:
        print("Found 7! Stopping.")
        break
    print(i)

print("Loop ended")
````

````python
# Find first even number in a list
numbers = [1, 3, 7, 8, 9, 12, 15]

for num in numbers:
    if num % 2 == 0:
        print(f"First even number: {num}")
        break
````

````python
# Number guessing game
secret_number = 7
attempts = 0
max_attempts = 5

while attempts < max_attempts:
    guess = int(input("Guess the number (1-10): "))
    attempts += 1
    
    if guess == secret_number:
        print(f"Correct! You won in {attempts} attempts!")
        break
    elif guess < secret_number:
        print("Too low!")
    else:
        print("Too high!")
else:
    print(f"Game over! The number was {secret_number}")
````

## Partie 6 : Instruction Continue - Passer l'Itération ⏭️

L'instruction **continue** saute le reste de l'itération actuelle et passe à la suivante.

````python
# Print only odd numbers (skip even numbers)
for i in range(1, 11):
    if i % 2 == 0:
        continue  # Skip the rest for even numbers
    print(i)
````

````python
# Skip negative numbers
numbers = [-5, 3, -2, 8, -1, 10, 15]

for num in numbers:
    if num < 0:
        continue  # Skip negative numbers
    print(num)
````

````python
# Skip multiples of 3
for i in range(1, 21):
    if i % 3 == 0:
        continue
    print(i, end=" ")
print()
````

````python
# Sum only positive numbers
numbers = [10, -5, 8, -3, 15, -7, 20]
total = 0

for num in numbers:
    if num < 0:
        continue  # Skip negatives
    total += num

print(f"Sum of positive numbers: {total}")
````

### Comparaison Break vs Continue

````python
# Using BREAK - stops completely at 5
print("With BREAK:")
for i in range(1, 11):
    if i == 5:
        break
    print(i, end=" ")
print()

# Using CONTINUE - skips only 5
print("\nWith CONTINUE:")
for i in range(1, 11):
    if i == 5:
        continue
    print(i, end=" ")
print()
````

## Partie 7 : Boucles Imbriquées - Boucles Dans des Boucles 🔄🔄

Vous pouvez mettre des boucles à l'intérieur d'autres boucles !

### Exemple de Boucle Imbriquée Basique

````python
# Outer loop runs 3 times
for i in range(1, 4):
    print(f"Outer loop: {i}")
    
    # Inner loop runs 3 times for EACH outer loop
    for j in range(1, 4):
        print(f"  Inner loop: {j}")
````

### Table de Multiplication

````python
# Full multiplication table (1-5)
for i in range(1, 6):
    for j in range(1, 6):
        result = i * j
        print(f"{i} x {j} = {result:2d}", end="  |  ")
    print()  # New line after each row
````

### Impression de Motifs

````python
# Rectangle pattern
for row in range(4):
    for col in range(6):
        print("*", end=" ")
    print()  # New line after each row
````

````python
# Triangle pattern
for row in range(1, 6):
    for col in range(row):
        print("*", end=" ")
    print()  # New line
````

````python
# Number triangle
for i in range(1, 6):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
````

````python
# Grid with coordinates
for row in range(3):
    for col in range(4):
        print(f"({row},{col})", end=" ")
    print()
````

### Boucle Imbriquée avec Listes

````python
# Compare each pair of numbers
numbers1 = [1, 2, 3]
numbers2 = [4, 5, 6]

for num1 in numbers1:
    for num2 in numbers2:
        print(f"{num1} + {num2} = {num1 + num2}")
````

## Partie 8 : Applications Réelles 🌍

Appliquons les boucles pour résoudre des problèmes pratiques !

### Application 1 : Calculateur de Moyenne

````python
# Calculate average of N numbers
n = int(input("How many numbers? "))
total = 0

for i in range(n):
    num = float(input(f"Enter number {i+1}: "))
    total += num

average = total / n
print(f"\nAverage: {average:.2f}")
````

### Application 2 : Calculateur de Factorielle

````python
# Calculate factorial (e.g., 5! = 5 × 4 × 3 × 2 × 1)
n = int(input("Enter a number: "))
factorial = 1

for i in range(1, n + 1):
    factorial *= i

print(f"{n}! = {factorial}")
````

### Application 3 : Vérificateur de Nombre Premier

````python
# Check if a number is prime
num = int(input("Enter a number: "))
is_prime = True

if num < 2:
    is_prime = False
else:
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break

if is_prime:
    print(f"{num} is PRIME")
else:
    print(f"{num} is NOT prime")
````

### Application 4 : Jeu FizzBuzz

````python
# FizzBuzz: Print numbers 1-30, but:
# - "Fizz" for multiples of 3
# - "Buzz" for multiples of 5
# - "FizzBuzz" for multiples of both

for i in range(1, 31):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz", end=" ")
    elif i % 3 == 0:
        print("Fizz", end=" ")
    elif i % 5 == 0:
        print("Buzz", end=" ")
    else:
        print(i, end=" ")
print()
````

### Application 5 : Vérificateur de Force de Mot de Passe

````python
# Check password strength
password = input("Enter a password: ")

has_upper = False
has_lower = False
has_digit = False

for char in password:
    if char.isupper():
        has_upper = True
    elif char.islower():
        has_lower = True
    elif char.isdigit():
        has_digit = True

if len(password) >= 8 and has_upper and has_lower and has_digit:
    print("Strong password! ✅")
else:
    print("Weak password! Need: 8+ chars, uppercase, lowercase, and digit")
````

### Application 6 : Suite de Fibonacci

````python
# Generate first N Fibonacci numbers
# Sequence: 0, 1, 1, 2, 3, 5, 8, 13, ...

n = int(input("How many Fibonacci numbers? "))
a, b = 0, 1

for i in range(n):
    print(a, end=" ")
    a, b = b, a + b  # Simultaneous assignment
print()
````

### Application 7 : Compteur de Fréquence de Caractères

````python
# Count frequency of each character
text = input("Enter text: ")

print("\nCharacter frequencies:")
for char in "abcdefghijklmnopqrstuvwxyz":
    count = 0
    for c in text.lower():
        if c == char:
            count += 1
    if count > 0:
        print(f"{char}: {count}")
````

## Partie 9 : Utiliser le Module random 🎲

Le **module random** fournit des outils pour générer des nombres aléatoires et faire des choix aléatoires - essentiel pour les jeux, simulations et tests !

### D'abord : Importer le Module

### random.randint(min, max) - Entiers Aléatoires

Génère un entier aléatoire entre min et max (inclusif aux deux extrémités) :

````python
import random

# Generate random number between 1 and 10
random_num = random.randint(1, 10)
print(f"Random number (1-10): {random_num}")

# Generate 5 random numbers
print("\n5 random numbers between 1-100:")
for i in range(5):
    print(random.randint(1, 100), end=" ")
print()
````

### random.choice(list) - Sélection Aléatoire

Choisit un élément aléatoire dans une liste :

````python
import random

# Choose a random color
colors = ["red", "blue", "green", "yellow", "purple"]
random_color = random.choice(colors)
print(f"Random color: {random_color}")

# Choose a random letter
letters = ['A', 'B', 'C', 'D', 'E']
print(f"Random letter: {random.choice(letters)}")

# Pick 3 random fruits (can repeat)
fruits = ["apple", "banana", "orange", "mango", "grape"]
print("\n3 random fruits:")
for i in range(3):
    print(f"  {i+1}. {random.choice(fruits)}")
````

### random.random() - Décimaux Aléatoires

Génère un nombre décimal aléatoire entre 0.0 et 1.0 :

````python
import random

# Generate random float between 0 and 1
random_float = random.random()
print(f"Random float (0-1): {random_float}")

# Use it for probability (50% chance)
if random.random() < 0.5:
    print("Heads!")
else:
    print("Tails!")

# Generate random float in a custom range (e.g., 0-100)
random_percent = random.random() * 100
print(f"Random percentage: {random_percent:.2f}%")
````

### Exemple 1 : Jeu de Lancer de Dés

Simulez le lancer de dés - parfait pour les jeux !

````python
import random

# Roll a single die (1-6)
die = random.randint(1, 6)
print(f"You rolled: {die}")

# Roll two dice
die1 = random.randint(1, 6)
die2 = random.randint(1, 6)
total = die1 + die2
print(f"\nDice 1: {die1}")
print(f"Dice 2: {die2}")
print(f"Total: {total}")

# Roll the dice 10 times
print("\nRolling 10 times:")
for i in range(10):
    roll = random.randint(1, 6)
    print(f"Roll {i+1}: {roll}")
````

### Exemple 2 : Simulateur de Pile ou Face

Simulez le tirage à pile ou face en utilisant des choix aléatoires :

````python
import random

# Single coin flip
coin = random.choice(["Heads", "Tails"])
print(f"Coin flip: {coin}")

# Flip 20 times and count results
heads_count = 0
tails_count = 0

print("\nFlipping 20 times:")
for i in range(20):
    flip = random.choice(["Heads", "Tails"])
    print(flip, end=" ")
    
    if flip == "Heads":
        heads_count += 1
    else:
        tails_count += 1

print(f"\n\nResults:")
print(f"Heads: {heads_count}")
print(f"Tails: {tails_count}")
````
