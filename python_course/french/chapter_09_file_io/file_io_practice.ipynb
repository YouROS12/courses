## Partie 1 : Pourquoi Travailler avec des Fichiers ?

Sans fichiers, les données de votre programme disparaissent quand il se ferme. Avec des fichiers, vos programmes peuvent :
- **Sauvegarder les données utilisateur** (paramètres, progression, notes)
- **Partager des données** avec d'autres programmes
- **Persister les informations** entre les sessions
- **Construire de vraies applications** que les gens utilisent réellement !

Pensez à vos applications préférées - elles utilisent toutes des fichiers pour se souvenir de vos données !

````python
# Without files: Data is lost
user_score = 1000
print(f"Your score: {user_score}")

# When the program closes, this score is GONE forever!
# You'd have to start from 0 every time.
````

````python
# With files: Data persists!
# We can save the score to a file and load it next time

# Saving (we'll learn the details soon)
score = 1000
with open("game_score.txt", "w") as file:
    file.write(str(score))

print(f"Score saved: {score}")
print("Even if we close the program, the score is saved in game_score.txt!")
````

````python
# Loading the score back (maybe days later)
with open("game_score.txt", "r") as file:
    saved_score = int(file.read())

print(f"Welcome back! Your score: {saved_score}")
print("The data persisted!")
````

## Partie 2 : Ouvrir et Fermer des Fichiers

Pour travailler avec un fichier, vous devez :
1. **L'ouvrir** (dire à Python quel fichier utiliser)
2. **Faire quelque chose** avec (lire, écrire, etc.)
3. **Le fermer** (très important !)

### Modes de Fichier :
- **"r"** - Lecture (par défaut) - le fichier doit exister
- **"w"** - Écriture - crée un nouveau fichier ou **écrase** l'existant
- **"a"** - Ajout - ajoute à la fin du fichier
- **"r+"** - Lecture et écriture

````python
# Basic file operations (old way - not recommended)
# We'll learn the better way soon!

# Create a file and write to it
file = open("test.txt", "w")  # Open in write mode
file.write("Hello, World!\n")
file.write("This is a test file.")
file.close()  # MUST close the file!

print("File created: test.txt")
````

````python
# Read the file we just created
file = open("test.txt", "r")  # Open in read mode
content = file.read()
file.close()  # MUST close!

print("File contents:")
print(content)
````

````python
# Why closing is important:
# - Frees up system resources
# - Ensures data is actually written to disk
# - Allows other programs to access the file
# - Prevents file corruption

# Problem: If your code crashes before file.close(), the file stays open!
# Solution: Use the 'with' statement (coming next!)
````

## Partie 3 : Lecture de Fichiers (read, readline, readlines)

Python vous donne trois façons de lire des fichiers :
1. **read()** - Lire le fichier entier comme une seule chaîne
2. **readline()** - Lire une ligne à la fois
3. **readlines()** - Lire toutes les lignes dans une liste

````python
# First, let's create a sample file with multiple lines
sample_content = """Line 1: Python is awesome
Line 2: Files are useful
Line 3: Let's learn file I/O
Line 4: This is fun!
Line 5: Keep coding!"""

with open("sample.txt", "w") as file:
    file.write(sample_content)

print("Created sample.txt")
````

````python
# Method 1: read() - Read entire file as one string
with open("sample.txt", "r") as file:
    content = file.read()

print("Using read():")
print(content)
print(f"\nType: {type(content)}")
print(f"Length: {len(content)} characters")
````

````python
# Method 2: readline() - Read one line at a time
with open("sample.txt", "r") as file:
    line1 = file.readline()
    line2 = file.readline()
    line3 = file.readline()

print("Using readline():")
print(f"First line: {line1.strip()}")  # .strip() removes \n
print(f"Second line: {line2.strip()}")
print(f"Third line: {line3.strip()}")
````

````python
# Method 3: readlines() - Read all lines into a list
with open("sample.txt", "r") as file:
    lines = file.readlines()

print("Using readlines():")
print(f"Type: {type(lines)}")
print(f"Number of lines: {len(lines)}")
print("\nAll lines:")
for i, line in enumerate(lines, 1):
    print(f"{i}. {line.strip()}")
````

````python
# Bonus: You can iterate over a file directly (most efficient!)
with open("sample.txt", "r") as file:
    print("Iterating directly over file:")
    for line in file:
        print(f"  - {line.strip()}")
````

## Partie 4 : Écrire dans des Fichiers

Écrire dans des fichiers vous permet de sauvegarder des données, créer des journaux, exporter des résultats, et plus encore !

**Important :** Le mode "w" **écrase** le fichier entier !

````python
# Writing to a file (creates new or overwrites existing)
with open("output.txt", "w") as file:
    file.write("Hello, World!\n")
    file.write("This is line 2\n")
    file.write("This is line 3\n")

print("File written: output.txt")
````

````python
# Read it back to verify
with open("output.txt", "r") as file:
    content = file.read()

print("Contents of output.txt:")
print(content)
````

````python
# Writing multiple lines with writelines()
lines = [
    "First line\n",
    "Second line\n",
    "Third line\n"
]

with open("multi_lines.txt", "w") as file:
    file.writelines(lines)

print("Multiple lines written!")

# Verify
with open("multi_lines.txt", "r") as file:
    print(file.read())
````

````python
# Writing numbers and other data types (must convert to string!)
numbers = [1, 2, 3, 4, 5]
total = sum(numbers)
average = total / len(numbers)

with open("stats.txt", "w") as file:
    file.write(f"Numbers: {numbers}\n")
    file.write(f"Total: {total}\n")
    file.write(f"Average: {average}\n")

print("Stats saved!")

# Read back
with open("stats.txt", "r") as file:
    print(file.read())
````

````python
# Demonstrating that "w" mode OVERWRITES
with open("overwrite_test.txt", "w") as file:
    file.write("Original content\n")

print("First write:")
with open("overwrite_test.txt", "r") as file:
    print(file.read())

# Write again - this REPLACES the content!
with open("overwrite_test.txt", "w") as file:
    file.write("New content (old content is gone!)\n")

print("Second write:")
with open("overwrite_test.txt", "r") as file:
    print(file.read())
````

## Partie 5 : Ajouter à des Fichiers

Le mode ajout ("a") ajoute à la fin d'un fichier **sans effacer** le contenu existant. Parfait pour les journaux !

````python
# Create initial file
with open("log.txt", "w") as file:
    file.write("=== LOG FILE ===\n")
    file.write("Entry 1: Program started\n")

print("Initial log:")
with open("log.txt", "r") as file:
    print(file.read())
````

````python
# Append to the file (preserves existing content)
with open("log.txt", "a") as file:
    file.write("Entry 2: User logged in\n")
    file.write("Entry 3: Data loaded\n")

print("After appending:")
with open("log.txt", "r") as file:
    print(file.read())
````

````python
# Append more entries
with open("log.txt", "a") as file:
    file.write("Entry 4: Processing complete\n")
    file.write("Entry 5: Program closed\n")

print("Final log:")
with open("log.txt", "r") as file:
    print(file.read())
````

````python
# Real-world example: Activity log
activities = [
    "Wake up",
    "Breakfast",
    "Study Python",
    "Lunch",
    "Practice coding"
]

# Create log
with open("activity_log.txt", "w") as file:
    file.write("=== MY DAILY ACTIVITIES ===\n\n")

# Append activities one by one
for i, activity in enumerate(activities, 1):
    with open("activity_log.txt", "a") as file:
        file.write(f"{i}. {activity}\n")

# Read the complete log
with open("activity_log.txt", "r") as file:
    print(file.read())
````

## Partie 6 : L'Instruction 'with' (Meilleure Pratique !)

L'instruction **with** est la bonne façon de travailler avec des fichiers. Elle :
- Ferme automatiquement les fichiers (même si des erreurs surviennent)
- Rend le code plus propre et sûr
- Est la **meilleure pratique Python**

**Utilisez toujours 'with' quand vous travaillez avec des fichiers !**

````python
# Old way (not recommended)
file = open("old_way.txt", "w")
file.write("This is the old way")
file.close()  # Must remember to close!

# Problem: If an error occurs before close(), file stays open!
print("Old way works, but risky")
````

````python
# Better way (recommended)
with open("better_way.txt", "w") as file:
    file.write("This is the better way")
    # File automatically closes when we exit the 'with' block
    # Even if an error occurs!

print("Better way - file automatically closed!")
````

````python
# The 'with' statement handles errors gracefully
try:
    with open("error_test.txt", "w") as file:
        file.write("Line 1\n")
        file.write("Line 2\n")
        # Imagine an error happens here
        # x = 1 / 0  # Uncomment to test
        file.write("Line 3\n")
    # File is STILL closed properly, even if error occurred!
except Exception as e:
    print(f"Error occurred: {e}")
    print("But the file was still closed safely!")

# Verify
with open("error_test.txt", "r") as file:
    print("\nFile contents:")
    print(file.read())
````

````python
# You can open multiple files at once with 'with'
with open("source.txt", "w") as source:
    source.write("Original data\n")
    source.write("Line 2\n")
    source.write("Line 3\n")

# Copy one file to another
with open("source.txt", "r") as source, open("destination.txt", "w") as dest:
    content = source.read()
    dest.write(content)

print("File copied!")

# Verify
with open("destination.txt", "r") as file:
    print(file.read())
````

### Point Clé :
**Utilisez toujours `with open(...) as file:` au lieu de `file = open(...)`**

C'est plus sûr, plus propre et la façon Python !

## Partie 7 : Chemins de Fichiers (Absolu vs Relatif)

Les chemins de fichiers indiquent à Python **où** trouver un fichier.

**Chemin relatif :** Depuis le dossier actuel
**Chemin absolu :** Chemin complet depuis la racine

````python
import os

# Check current working directory
current_dir = os.getcwd()
print(f"Current directory: {current_dir}")
````

````python
# Relative paths (from current directory)

# Same folder
with open("relative_example.txt", "w") as file:
    file.write("Using relative path - same folder")

print("Created: relative_example.txt (in current folder)")
````

````python
# Create a subfolder and file
os.makedirs("data", exist_ok=True)  # Create folder if it doesn't exist

# Subfolder
with open("data/info.txt", "w") as file:
    file.write("File in subfolder")

print("Created: data/info.txt (in subfolder)")
````

````python
# Read from subfolder
with open("data/info.txt", "r") as file:
    content = file.read()

print(f"Content: {content}")
````

````python
# Absolute paths (complete path from root)
# These vary by operating system

# Get absolute path of our file
absolute_path = os.path.abspath("relative_example.txt")
print(f"Absolute path: {absolute_path}")

# You could open it like this (but relative is usually better)
with open(absolute_path, "r") as file:
    print(f"Content: {file.read()}")
````

````python
# Useful path operations
print(f"Does file exist? {os.path.exists('relative_example.txt')}")
print(f"Is it a file? {os.path.isfile('relative_example.txt')}")
print(f"Is it a directory? {os.path.isdir('data')}")
print(f"File size: {os.path.getsize('relative_example.txt')} bytes")
````

### Meilleures Pratiques :
- Utilisez des **chemins relatifs** pour un code portable
- Utilisez le module **os.path** pour les opérations de chemin
- Utilisez des **slashes obliques** (/) - fonctionne sur tous les systèmes
- Vérifiez si le fichier existe avant de lire : `os.path.exists()`

## Partie 8 : Travailler avec des Fichiers CSV

Les fichiers **CSV (Comma-Separated Values)** sont parfaits pour des données type tableur.

Format :
```
Name,Age,Grade
Alice,20,A
Bob,19,B
```

````python
import csv

# Writing CSV files
students = [
    ["Name", "Age", "Grade"],
    ["Alice", 20, "A"],
    ["Bob", 19, "B"],
    ["Carol", 21, "A"],
    ["David", 20, "C"]
]

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(students)  # Write all rows at once

print("Created students.csv")
````

````python
# Reading CSV files
with open("students.csv", "r") as file:
    reader = csv.reader(file)
    
    print("Students from CSV:")
    for row in reader:
        print(row)
````

````python
# Reading with headers
with open("students.csv", "r") as file:
    reader = csv.reader(file)
    
    header = next(reader)  # Get first row (header)
    print(f"Headers: {header}\n")
    
    print("Student data:")
    for row in reader:
        name, age, grade = row
        print(f"  {name} is {age} years old with grade {grade}")
````

````python
# Using DictReader (easier to work with)
with open("students.csv", "r") as file:
    reader = csv.DictReader(file)
    
    print("Using DictReader:")
    for row in reader:
        print(f"{row['Name']} - Age: {row['Age']}, Grade: {row['Grade']}")
````

````python
# Writing with DictWriter
students_dict = [
    {"Name": "Eve", "Age": 22, "Grade": "A", "City": "New York"},
    {"Name": "Frank", "Age": 20, "Grade": "B", "City": "Boston"},
    {"Name": "Grace", "Age": 21, "Grade": "A", "City": "Chicago"}
]

with open("students_detailed.csv", "w", newline="") as file:
    fieldnames = ["Name", "Age", "Grade", "City"]
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    
    writer.writeheader()  # Write header row
    writer.writerows(students_dict)  # Write data rows

print("Created students_detailed.csv")

# Read it back
with open("students_detailed.csv", "r") as file:
    print("\nContents:")
    print(file.read())
````

## Partie 9 : Travailler avec des Fichiers JSON

**JSON (JavaScript Object Notation)** est parfait pour stocker des dictionnaires, listes et données structurées.

Dictionnaires Python ↔ Fichiers JSON

````python
import json

# Python dictionary
user_data = {
    "username": "alice",
    "age": 25,
    "email": "alice@example.com",
    "interests": ["Python", "Data Science", "Reading"],
    "is_active": True
}

# Save to JSON file
with open("user.json", "w") as file:
    json.dump(user_data, file, indent=4)  # indent=4 makes it pretty

print("Created user.json")
````

````python
# Let's see what it looks like
with open("user.json", "r") as file:
    print("Contents of user.json:")
    print(file.read())
````

````python
# Load JSON file back to Python
with open("user.json", "r") as file:
    loaded_data = json.load(file)

print(f"Type: {type(loaded_data)}")
print(f"Username: {loaded_data['username']}")
print(f"Age: {loaded_data['age']}")
print(f"Interests: {loaded_data['interests']}")
````

````python
# Saving complex data structures
game_data = {
    "player": {
        "name": "Hero",
        "level": 15,
        "health": 100,
        "inventory": ["sword", "shield", "potion", "map"]
    },
    "settings": {
        "difficulty": "hard",
        "sound": True,
        "music_volume": 75
    },
    "highscores": [1500, 1200, 1000, 800, 500]
}

with open("game_save.json", "w") as file:
    json.dump(game_data, file, indent=2)

print("Game data saved!")
````

````python
# Load game data
with open("game_save.json", "r") as file:
    loaded_game = json.load(file)

print("Game loaded!")
print(f"Player: {loaded_game['player']['name']}")
print(f"Level: {loaded_game['player']['level']}")
print(f"Inventory: {loaded_game['player']['inventory']}")
print(f"Difficulty: {loaded_game['settings']['difficulty']}")
````

````python
# Converting between JSON strings and Python objects
data = {"name": "Bob", "score": 95}

# Python dict → JSON string
json_string = json.dumps(data)
print(f"JSON string: {json_string}")
print(f"Type: {type(json_string)}")

# JSON string → Python dict
parsed_data = json.loads(json_string)
print(f"\nParsed data: {parsed_data}")
print(f"Type: {type(parsed_data)}")
````

Ajoutez des horodatages à vos fichiers pour suivre quand les événements se produisent !

````python
from datetime import datetime

# Get current date and time
now = datetime.now()

print(f"Current datetime: {now}")
print(f"Type: {type(now)}")
````

````python
# Formatting timestamps
now = datetime.now()

# Different formats
format1 = now.strftime("%Y-%m-%d %H:%M:%S")
format2 = now.strftime("%B %d, %Y")
format3 = now.strftime("%I:%M %p")
format4 = now.strftime("%A, %B %d, %Y at %I:%M %p")

print(f"Standard: {format1}")
print(f"Date only: {format2}")
print(f"Time only: {format3}")
print(f"Full: {format4}")
````

````python
# Common datetime format codes:
print("""
Format Codes:
  %Y - Year (4 digits)      %m - Month (01-12)        %d - Day (01-31)
  %H - Hour (00-23)         %M - Minute (00-59)       %S - Second (00-59)
  %I - Hour (01-12)         %p - AM/PM
  %A - Weekday (full)       %B - Month name (full)
  %a - Weekday (abbr)       %b - Month name (abbr)
""")
````

````python
# Creating a timestamped log file
from datetime import datetime

def log_event(message):
    """Add a timestamped entry to the log file."""
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    with open("app_log.txt", "a") as file:
        file.write(f"[{timestamp}] {message}\n")

# Create log entries
log_event("Application started")
log_event("User logged in: alice")
log_event("Data loaded successfully")
log_event("User performed search query")
log_event("Application closed")

print("Log file created!")
````

````python
# Read the log file
with open("app_log.txt", "r") as file:
    print("Application Log:")
    print(file.read())
````

````python
# Creating a daily log file (different file each day)
from datetime import datetime

# Generate filename with today's date
today = datetime.now().strftime("%Y-%m-%d")
log_filename = f"log_{today}.txt"

with open(log_filename, "a") as file:
    timestamp = datetime.now().strftime("%H:%M:%S")
    file.write(f"[{timestamp}] Daily log entry\n")

print(f"Created: {log_filename}")
````

````python
# Saving data with timestamp
import json
from datetime import datetime

data = {
    "user": "alice",
    "action": "completed_quiz",
    "score": 95,
    "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
}

with open("user_activity.json", "w") as file:
    json.dump(data, file, indent=2)

print("Activity saved with timestamp!")

# Read it back
with open("user_activity.json", "r") as file:
    loaded = json.load(file)
    print(f"\nUser: {loaded['user']}")
    print(f"Action: {loaded['action']}")
    print(f"Time: {loaded['timestamp']}")
````

### Application 1 : Application Simple de Prise de Notes

````python
from datetime import datetime

class NoteApp:
    def __init__(self, filename="notes.txt"):
        self.filename = filename
    
    def add_note(self, note):
        """Add a new note with timestamp."""
        timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        with open(self.filename, "a") as file:
            file.write(f"[{timestamp}] {note}\n")
        print(f"Note added!")
    
    def view_notes(self):
        """Display all notes."""
        try:
            with open(self.filename, "r") as file:
                notes = file.read()
                if notes:
                    print("=== YOUR NOTES ===")
                    print(notes)
                else:
                    print("No notes yet!")
        except FileNotFoundError:
            print("No notes yet!")
    
    def clear_notes(self):
        """Delete all notes."""
        with open(self.filename, "w") as file:
            file.write("")
        print("All notes cleared!")

# Test the note app
app = NoteApp()
app.add_note("Remember to study Python every day")
app.add_note("Buy groceries: milk, bread, eggs")
app.add_note("Call mom this weekend")
app.view_notes()
````

### Application 2 : Gestionnaire de Contacts avec Persistance

````python
import json

class ContactManager:
    def __init__(self, filename="contacts.json"):
        self.filename = filename
        self.contacts = self.load_contacts()
    
    def load_contacts(self):
        """Load contacts from file."""
        try:
            with open(self.filename, "r") as file:
                return json.load(file)
        except FileNotFoundError:
            return {}
    
    def save_contacts(self):
        """Save contacts to file."""
        with open(self.filename, "w") as file:
            json.dump(self.contacts, file, indent=2)
    
    def add_contact(self, name, phone, email):
        """Add a new contact."""
        self.contacts[name] = {
            "phone": phone,
            "email": email
        }
        self.save_contacts()
        print(f"Contact '{name}' added!")
    
    def get_contact(self, name):
        """Get a contact by name."""
        if name in self.contacts:
            contact = self.contacts[name]
            print(f"\nName: {name}")
            print(f"Phone: {contact['phone']}")
            print(f"Email: {contact['email']}")
        else:
            print(f"Contact '{name}' not found!")
    
    def list_contacts(self):
        """List all contacts."""
        if self.contacts:
            print("\n=== ALL CONTACTS ===")
            for name, info in self.contacts.items():
                print(f"\n{name}:")
                print(f"  Phone: {info['phone']}")
                print(f"  Email: {info['email']}")
        else:
            print("No contacts yet!")

# Test the contact manager
manager = ContactManager()
manager.add_contact("Alice", "555-0101", "alice@email.com")
manager.add_contact("Bob", "555-0102", "bob@email.com")
manager.add_contact("Carol", "555-0103", "carol@email.com")
manager.list_contacts()
````

### Application 3 : Enregistreur d'Activités

````python
from datetime import datetime

class ActivityLogger:
    def __init__(self, filename="activity.log"):
        self.filename = filename
    
    def log(self, level, message):
        """Log a message with timestamp and level."""
        timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        with open(self.filename, "a") as file:
            file.write(f"[{timestamp}] [{level.upper()}] {message}\n")
    
    def info(self, message):
        """Log info message."""
        self.log("INFO", message)
    
    def warning(self, message):
        """Log warning message."""
        self.log("WARNING", message)
    
    def error(self, message):
        """Log error message."""
        self.log("ERROR", message)
    
    def view_log(self):
        """Display the entire log."""
        try:
            with open(self.filename, "r") as file:
                print(file.read())
        except FileNotFoundError:
            print("No log entries yet!")

# Test the logger
logger = ActivityLogger()
logger.info("Application started")
logger.info("User 'alice' logged in")
logger.warning("Database connection slow")
logger.info("Data loaded successfully")
logger.error("Failed to save user preferences")
logger.info("Application closed")

print("\n=== LOG FILE ===")
logger.view_log()
````

### Référence Rapide :

**Lecture de Fichiers :**
```python
# Lire le fichier entier
with open("file.txt", "r") as file:
    content = file.read()

# Lire ligne par ligne
with open("file.txt", "r") as file:
    for line in file:
        print(line.strip())
```

**Écriture de Fichiers :**
```python
# Écrire (écrase)
with open("file.txt", "w") as file:
    file.write("Hello, World!\n")

# Ajouter (ajoute à la fin)
with open("file.txt", "a") as file:
    file.write("Nouvelle ligne\n")
```

**Fichiers CSV :**
```python
import csv

# Écrire CSV
with open("data.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Nom", "Âge"])

# Lire CSV
with open("data.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

**Fichiers JSON :**
```python
import json

# Sauvegarder en JSON
data = {"nom": "Alice", "age": 25}
with open("data.json", "w") as file:
    json.dump(data, file, indent=2)

# Charger depuis JSON
with open("data.json", "r") as file:
    data = json.load(file)
```
