## Sample Solutions for Challenges

### Challenge 1: Safe Division - Sample Solution
```python
def safe_divide(a, b):
    try:
        result = a / b
        return result
    except ZeroDivisionError:
        return "Cannot divide by zero"
    except TypeError:
        return "Invalid types for division"

# Test cases
print(safe_divide(10, 2))      # Returns 5.0
print(safe_divide(10, 0))      # Returns "Cannot divide by zero"
print(safe_divide("10", 2))    # Returns "Invalid types for division"
```

### Challenge 2: File Reader - Sample Solution
```python
def read_file_safely(filename):
    try:
        with open(filename, "r") as file:
            content = file.read()
            return content
    except FileNotFoundError:
        return "File not found"
    finally:
        print("Operation complete")

# Test with existing file
with open("test.txt", "w") as f:
    f.write("Hello, World!")

print(read_file_safely("test.txt"))          # Returns content
print(read_file_safely("nonexistent.txt"))   # Returns "File not found"
```

### Challenge 3: Input Validator - Sample Solution
```python
def validate_age(age_str):
    try:
        age = int(age_str)
        
        if age < 0:
            raise ValueError("Age cannot be negative")
        
        if age > 150:
            raise ValueError("Age cannot be over 150")
        
        return age
    
    except ValueError as e:
        return f"Error: {e}"

# Test cases
print(validate_age("25"))    # Returns 25
print(validate_age("-5"))    # Returns "Error: Age cannot be negative"
print(validate_age("200"))   # Returns "Error: Age cannot be over 150"
print(validate_age("abc"))   # Returns "Error: invalid literal for int()..."
```

### Challenge 4: Student Grade Manager - Sample Solution
```python
def calculate_average(scores):
    try:
        if len(scores) == 0:
            return 0
        return sum(scores) / len(scores)
    except TypeError:
        print("Error: All scores must be numeric")
        return 0

def get_letter_grade(average):
    if average >= 90:
        return "A"
    elif average >= 80:
        return "B"
    elif average >= 70:
        return "C"
    elif average >= 60:
        return "D"
    else:
        return "F"

# Test data
students = {
    "Alice": [95, 87, 92, 88],
    "Bob": [78, 82, 75, 80],
    "Charlie": [65, 70, 68, 72]
}

# Process each student
for name, scores in students.items():
    try:
        average = calculate_average(scores)
        letter = get_letter_grade(average)
        print(f"{name}: Average = {average:.2f}, Grade = {letter}")
    except Exception as e:
        print(f"Error processing {name}: {e}")
```
