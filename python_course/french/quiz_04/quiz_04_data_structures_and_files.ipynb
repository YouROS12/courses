### Défi 4 : Notes d'étudiants - Exemple de solution
```python
# Créer le dictionnaire de notes
grades = {"Alice": 92, "Bob": 87, "Charlie": 95, "Diana": 88}

# Calculer la moyenne
average = sum(grades.values()) / len(grades)
print("Score moyen:", average)

# Trouver le score le plus élevé
highest_score = max(grades.values())
for student, score in grades.items():
    if score == highest_score:
        print(f"Le plus élevé: {student} avec {score}")
        break

# Trouver le score le plus bas
lowest_score = min(grades.values())
for student, score in grades.items():
    if score == lowest_score:
        print(f"Le plus bas: {student} avec {score}")
        break

# Afficher tous les étudiants avec leurs scores
print("\nTous les étudiants:")
for student, score in grades.items():
    print(f"{student}: {score}")
```
