## Partie 1 : Comprendre les Erreurs et Exceptions

**Les erreurs sont inévitables en programmation** - même les développeurs expérimentés les rencontrent quotidiennement !

Python a deux types d'erreurs :
1. **Erreurs de syntaxe** - Le code est mal écrit (Python ne peut même pas l'exécuter)
2. **Exceptions** - Le code est syntaxiquement correct, mais quelque chose ne va pas pendant l'exécution

**La gestion des exceptions** vous permet d'attraper ces erreurs d'exécution et de les gérer avec élégance au lieu de planter !

````python
# Example: What happens WITHOUT error handling
# Uncomment to see the crash:

# number = int("hello")  # This will crash!
# print("This line never runs")

print("Commented out to prevent crash!")
````

````python
# Example: WITH error handling (program continues!)
try:
    number = int("hello")
except ValueError:
    print("Oops! That's not a valid number.")

print("This line DOES run - program continues!")
````

````python
# Real-world example: User input
# Without error handling, one wrong input crashes everything

# Simulating user input
user_input = "twenty"  # User types this instead of "20"

try:
    age = int(user_input)
    print(f"You are {age} years old.")
except ValueError:
    print(f"'{user_input}' is not a valid number. Please enter digits only.")
````

## Partie 2 : Types d'Exceptions Courants

Python a de nombreux types d'exceptions intégrés. Voici ceux que vous rencontrerez le plus souvent :

### ValueError - Valeur Invalide pour une Opération

````python
# ValueError: Wrong value type for conversion
try:
    number = int("abc")  # Can't convert "abc" to integer
except ValueError as e:
    print(f"ValueError caught: {e}")
    print("This happens when converting invalid strings to numbers")
````

````python
# More ValueError examples
try:
    number = float("hello")
except ValueError:
    print("Can't convert 'hello' to float!")

try:
    number = int("12.5")  # int() expects no decimal point
except ValueError:
    print("int() can't convert strings with decimal points directly!")
````

### TypeError - Type Incorrect pour une Opération

````python
# TypeError: Incompatible types
try:
    result = "5" + 3  # Can't add string and integer
except TypeError as e:
    print(f"TypeError caught: {e}")
    print("You need to convert types first!")
    result = int("5") + 3
    print(f"Correct way: {result}")
````

````python
# More TypeError examples
try:
    result = len(5)  # len() doesn't work on integers
except TypeError:
    print("Can't get length of an integer!")

try:
    result = "hello"[1.5]  # Indices must be integers
except TypeError:
    print("String indices must be integers, not floats!")
````

### ZeroDivisionError - Division par Zéro

````python
# ZeroDivisionError: Dividing by zero
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero!")
    print("Mathematics doesn't allow this operation.")
````

````python
# Practical example: Safe division function
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Cannot divide by zero"

print(f"10 / 2 = {safe_divide(10, 2)}")
print(f"10 / 0 = {safe_divide(10, 0)}")
print(f"15 / 3 = {safe_divide(15, 3)}")
````

### FileNotFoundError - Fichier Inexistant

````python
# FileNotFoundError: Trying to open a file that doesn't exist
try:
    with open("nonexistent_file.txt", "r") as file:
        content = file.read()
except FileNotFoundError:
    print("File not found!")
    print("Make sure the file exists and the path is correct.")
````

````python
# Better approach: Check and provide helpful message
import os

filename = "data.txt"

try:
    with open(filename, "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print(f"Error: '{filename}' not found in {os.getcwd()}")
    print("Creating a new file instead...")
    with open(filename, "w") as file:
        file.write("New file created!")
    print(f"Created {filename} successfully!")
````

### IndexError - Index de Liste Hors Limites

````python
# IndexError: Accessing index that doesn't exist
numbers = [1, 2, 3, 4, 5]

try:
    print(numbers[10])  # List only has 5 elements (indices 0-4)
except IndexError:
    print(f"IndexError! List has {len(numbers)} elements (indices 0-{len(numbers)-1})")
    print(f"You tried to access index 10")
````

````python
# Safe way to access list elements
def safe_get_item(lst, index):
    try:
        return lst[index]
    except IndexError:
        return f"Index {index} out of range (list has {len(lst)} items)"

fruits = ["apple", "banana", "cherry"]
print(safe_get_item(fruits, 1))  # Works
print(safe_get_item(fruits, 10))  # Handles gracefully
````

### KeyError - Clé de Dictionnaire Introuvable

````python
# KeyError: Accessing key that doesn't exist in dictionary
student = {
    "name": "Alice",
    "age": 20,
    "grade": "A"
}

try:
    print(student["email"])  # 'email' key doesn't exist
except KeyError:
    print("KeyError! That key doesn't exist in the dictionary.")
    print(f"Available keys: {list(student.keys())}")
````

````python
# Better approach: Use .get() method (returns None if key doesn't exist)
email = student.get("email")
if email:
    print(f"Email: {email}")
else:
    print("Email not available")

# Or provide a default value
email = student.get("email", "not_provided@example.com")
print(f"Email: {email}")
````

## Partie 4 : Attraper des Exceptions Spécifiques

**Meilleure Pratique :** Attrapez toujours des exceptions **spécifiques**, pas toutes les exceptions !

Pourquoi ?
- Savoir exactement ce qui s'est mal passé
- Fournir des messages d'erreur appropriés
- Ne pas cacher les erreurs inattendues
- Faciliter le débogage

````python
# BAD: Catching all exceptions (too broad!)
try:
    number = int("hello")
except:  # This catches EVERYTHING - bad practice!
    print("Something went wrong")
    # But what? We don't know!
````

````python
# GOOD: Catching specific exceptions
try:
    number = int("hello")
except ValueError:  # Specific exception!
    print("ValueError: The input is not a valid number")
    # We know EXACTLY what went wrong
````

````python
# Getting error details with 'as'
try:
    result = 100 / 0
except ZeroDivisionError as error:
    print(f"Error type: {type(error).__name__}")
    print(f"Error message: {error}")
    print("This helps with debugging!")
````

````python
# Real-world example: Processing user data
def calculate_average(numbers_str):
    """Calculate average from comma-separated string of numbers."""
    try:
        # Split and convert to floats
        numbers = [float(n.strip()) for n in numbers_str.split(",")]
        average = sum(numbers) / len(numbers)
        return average
    except ValueError as e:
        return f"Error: Invalid number format - {e}"
    except ZeroDivisionError:
        return "Error: No numbers provided"

# Test it
print(calculate_average("10, 20, 30, 40"))  # Works
print(calculate_average("10, twenty, 30"))  # ValueError
print(calculate_average(""))  # ZeroDivisionError
````

## Partie 5 : Blocs Except Multiples

Vous pouvez avoir plusieurs blocs except pour gérer différentes exceptions différemment !

````python
# Multiple except blocks
def process_input(value):
    try:
        number = int(value)
        result = 100 / number
        return f"Result: {result}"
    except ValueError:
        return "Error: Please enter a valid integer"
    except ZeroDivisionError:
        return "Error: Cannot divide by zero"

# Test different scenarios
print(process_input("20"))     # Success
print(process_input("hello"))  # ValueError
print(process_input("0"))      # ZeroDivisionError
````

````python
# Catching multiple exceptions in one block (if same handling)
def parse_data(value):
    try:
        # Try to convert to int, float, or keep as string
        return int(value)
    except (ValueError, TypeError):
        # Both exceptions handled the same way
        return None

print(parse_data("42"))    # Returns 42
print(parse_data("hello")) # Returns None
print(parse_data(None))    # Returns None
````

````python
# Real-world example: Reading and processing a file
def read_and_sum_numbers(filename):
    """Read numbers from file and return their sum."""
    try:
        with open(filename, "r") as file:
            lines = file.readlines()
            
        numbers = [int(line.strip()) for line in lines]
        total = sum(numbers)
        return f"Sum: {total}"
        
    except FileNotFoundError:
        return f"Error: File '{filename}' not found"
    except ValueError as e:
        return f"Error: File contains invalid numbers - {e}"
    except PermissionError:
        return f"Error: No permission to read '{filename}'"

# Create a test file
with open("numbers.txt", "w") as f:
    f.write("10\n20\n30\n40\n50")

print(read_and_sum_numbers("numbers.txt"))  # Works
print(read_and_sum_numbers("missing.txt"))  # FileNotFoundError
````

````python
# Complex example: Dictionary operations with error handling
def get_student_grade(students, student_id, subject):
    """Get a student's grade for a subject."""
    try:
        student = students[student_id]
        grade = student[subject]
        return f"Grade: {grade}"
    except KeyError as e:
        return f"Error: Key {e} not found"
    except TypeError:
        return "Error: Invalid data structure"

# Test data
students_db = {
    "S001": {"name": "Alice", "math": 95, "science": 88},
    "S002": {"name": "Bob", "math": 87, "science": 92}
}

print(get_student_grade(students_db, "S001", "math"))     # Works
print(get_student_grade(students_db, "S999", "math"))     # Student not found
print(get_student_grade(students_db, "S001", "history"))  # Subject not found
````

## Partie 6 : La Clause else

La clause **else** s'exécute **uniquement si aucune exception ne s'est produite** dans le bloc try.

**Quand l'utiliser :**
- Code qui ne devrait s'exécuter qu'en cas de succès
- Séparer la gestion des erreurs de la logique de succès
- Rendre le code plus lisible

````python
# Basic else clause
try:
    age = int("25")
except ValueError:
    print("Invalid input!")
else:
    # This runs ONLY if no exception occurred
    print(f"Success! Age is {age}")
    print("Processing age...")
    if age >= 18:
        print("You are an adult.")
````

````python
# Comparing with and without else
# Without else (less clear)
try:
    number = int("42")
    print(f"Number: {number}")  # Runs even if exception might occur later
except ValueError:
    print("Error!")
````

````python
# With else (clearer intent)
try:
    number = int("42")
except ValueError:
    print("Error!")
else:
    # Clearly shows this is the "success" path
    print(f"Number: {number}")
    print("All good!")
````

````python
# Real-world example: File processing
def process_config_file(filename):
    try:
        with open(filename, "r") as file:
            config = file.read()
    except FileNotFoundError:
        print(f"Error: {filename} not found")
        print("Using default configuration")
        return {"mode": "default"}
    else:
        # File loaded successfully - process it
        print(f"Successfully loaded {filename}")
        print("Parsing configuration...")
        # Parse config here
        return {"mode": "custom", "data": config}

# Create test file
with open("config.txt", "w") as f:
    f.write("setting1=value1\nsetting2=value2")

result = process_config_file("config.txt")
print(f"Result: {result}")
````

````python
# Example: User login with validation
def login(username, password):
    # Simulated user database
    users = {
        "alice": "password123",
        "bob": "secret456"
    }
    
    try:
        stored_password = users[username]
    except KeyError:
        print(f"Login failed: Username '{username}' not found")
        return False
    else:
        # Username exists - now check password
        if password == stored_password:
            print(f"Welcome, {username}!")
            return True
        else:
            print("Login failed: Incorrect password")
            return False

# Test
login("alice", "password123")  # Success
login("alice", "wrong")        # Wrong password
login("charlie", "anything")   # User not found
````

## Partie 7 : La Clause finally

La clause **finally** s'exécute **TOUJOURS** - qu'une exception se soit produite ou non !

**Parfait pour :**
- Fermer des fichiers
- Libérer des ressources
- Opérations de nettoyage
- Journalisation

````python
# Basic finally clause
try:
    number = int("42")
    print(f"Number: {number}")
except ValueError:
    print("Error occurred!")
finally:
    # This ALWAYS runs - error or not!
    print("Finally block executed")
````

````python
# Demonstrating finally runs even with errors
try:
    result = 10 / 0  # This will fail
    print("This won't print")
except ZeroDivisionError:
    print("Error: Division by zero")
finally:
    print("Finally: This ALWAYS runs")
````

````python
# Real-world example: Resource cleanup
def process_file_with_cleanup(filename):
    file = None
    try:
        print(f"Opening {filename}...")
        file = open(filename, "r")
        content = file.read()
        print(f"Content: {content[:50]}...")  # First 50 chars
    except FileNotFoundError:
        print(f"Error: {filename} not found")
    finally:
        # Always close the file if it was opened
        if file:
            file.close()
            print("File closed successfully")
        print("Cleanup complete")

# Create test file
with open("test_data.txt", "w") as f:
    f.write("This is test data for the finally clause example.")

process_file_with_cleanup("test_data.txt")
````

````python
# Complete example: try-except-else-finally together!
def robust_division(a, b):
    print(f"Attempting to divide {a} by {b}")
    
    try:
        result = a / b
    except ZeroDivisionError:
        print("Error: Cannot divide by zero!")
        return None
    except TypeError:
        print("Error: Invalid types for division")
        return None
    else:
        # Runs only if no exception
        print(f"Success! Result: {result}")
        return result
    finally:
        # Always runs
        print("Operation complete\n")

# Test all scenarios
robust_division(10, 2)      # Success
robust_division(10, 0)      # ZeroDivisionError
robust_division("10", "2")  # TypeError
````

````python
# Practical example: Logging operations
from datetime import datetime

def perform_operation(operation_name, func):
    """Execute an operation with logging."""
    start_time = datetime.now()
    
    try:
        print(f"Starting: {operation_name}")
        result = func()
        return result
    except Exception as e:
        print(f"Error in {operation_name}: {e}")
        return None
    finally:
        end_time = datetime.now()
        duration = (end_time - start_time).total_seconds()
        print(f"Completed: {operation_name}")
        print(f"Duration: {duration:.4f} seconds\n")

# Test it
def successful_task():
    return sum(range(1000))

def failing_task():
    return 10 / 0

perform_operation("Calculate Sum", successful_task)
perform_operation("Divide by Zero", failing_task)
````

## Partie 8 : Lever des Exceptions

Vous pouvez **lever** des exceptions dans votre propre code quand quelque chose ne va pas !

**Quand lever des exceptions :**
- Entrée invalide pour vos fonctions
- Conditions qui ne devraient pas se produire
- Préconditions non respectées
- Pour signaler des problèmes au code appelant

````python
# Basic raise syntax
def check_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative!")
    if age > 150:
        raise ValueError("Age seems unrealistic!")
    print(f"Age {age} is valid")

# Test it
try:
    check_age(25)   # OK
    check_age(-5)   # Raises ValueError
except ValueError as e:
    print(f"Error: {e}")
````

````python
# Validating function inputs
def calculate_discount(price, discount_percent):
    """Calculate discounted price."""
    if price < 0:
        raise ValueError(f"Price cannot be negative: {price}")
    
    if discount_percent < 0 or discount_percent > 100:
        raise ValueError(f"Discount must be 0-100, got {discount_percent}")
    
    discount = price * (discount_percent / 100)
    return price - discount

# Valid usage
print(f"$100 with 20% off: ${calculate_discount(100, 20)}")

# Invalid usage
try:
    calculate_discount(100, 150)  # Invalid discount
except ValueError as e:
    print(f"Error: {e}")
````

````python
# Raising different exception types
def divide_numbers(a, b):
    """Divide two numbers with validation."""
    if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
        raise TypeError("Both arguments must be numbers")
    
    if b == 0:
        raise ZeroDivisionError("Cannot divide by zero")
    
    return a / b

# Test different error conditions
try:
    print(divide_numbers(10, 2))      # Works
    print(divide_numbers(10, 0))      # ZeroDivisionError
except (TypeError, ZeroDivisionError) as e:
    print(f"Error: {e}")

try:
    print(divide_numbers("10", "2"))  # TypeError
except (TypeError, ZeroDivisionError) as e:
    print(f"Error: {e}")
````

````python
# Real-world example: User registration validation
def register_user(username, password, email):
    """Register a new user with validation."""
    # Username validation
    if len(username) < 3:
        raise ValueError("Username must be at least 3 characters")
    if not username.isalnum():
        raise ValueError("Username must contain only letters and numbers")
    
    # Password validation
    if len(password) < 8:
        raise ValueError("Password must be at least 8 characters")
    
    # Email validation (basic)
    if "@" not in email or "." not in email:
        raise ValueError("Invalid email format")
    
    # If all validations pass
    return f"User '{username}' registered successfully!"

# Test valid registration
try:
    result = register_user("alice123", "securePass123", "alice@email.com")
    print(result)
except ValueError as e:
    print(f"Registration failed: {e}")

# Test invalid registrations
test_cases = [
    ("al", "password123", "alice@email.com"),           # Username too short
    ("alice", "pass", "alice@email.com"),               # Password too short
    ("alice123", "password123", "invalid-email"),       # Invalid email
]

for username, password, email in test_cases:
    try:
        register_user(username, password, email)
    except ValueError as e:
        print(f"Validation error: {e}")
````

## Partie 9 : Meilleures Pratiques pour la Gestion des Erreurs

Suivez ces meilleures pratiques pour écrire du code professionnel et maintenable !

### 1. Attraper des Exceptions Spécifiques (Pas Except Nu)

````python
# BAD: Too broad
try:
    number = int("hello")
except:  # Catches everything!
    print("Error")

# GOOD: Specific
try:
    number = int("hello")
except ValueError:  # Only catches ValueError
    print("Invalid number format")
````

### 2. Fournir des Messages d'Erreur Utiles

````python
# BAD: Vague message
try:
    age = int(input("Age: "))
except ValueError:
    print("Error")  # Not helpful!

# GOOD: Specific, helpful message
try:
    age = int(input("Age: "))
except ValueError:
    print("Please enter a valid number for age (digits only, no letters)")
````

### 3. N'Attrapez Pas les Exceptions Que Vous Ne Pouvez Pas Gérer

````python
# BAD: Hiding errors
try:
    # Some critical operation
    result = perform_critical_operation()
except Exception:
    pass  # Silently ignore - BAD!

# GOOD: Let critical errors propagate
try:
    result = perform_critical_operation()
except ValueError:
    # Only catch specific errors you can handle
    print("Invalid input")
# Other exceptions will raise - which is good!
````

### 4. Gérer les Erreurs Près de Leur Source

````python
# GOOD: Handle where it happens
def get_user_age():
    """Get age from user with validation."""
    try:
        age = int(input("Enter age: "))
        return age
    except ValueError:
        print("Invalid age entered")
        return None

# The error is handled right where it can occur
````

### 5. Utilisez try-except pour les Exceptions, Pas le Flux de Contrôle

````python
# BAD: Using exceptions for normal logic
try:
    user = users[username]
except KeyError:
    user = create_new_user(username)  # Not really an error!

# GOOD: Use normal conditional logic
if username in users:
    user = users[username]
else:
    user = create_new_user(username)
````

### 6. Documentez les Exceptions Que Vos Fonctions Lèvent

````python
def divide(a, b):
    """
    Divide two numbers.
    
    Args:
        a: The numerator (number)
        b: The denominator (number)
    
    Returns:
        float: The result of a / b
    
    Raises:
        TypeError: If a or b are not numbers
        ZeroDivisionError: If b is zero
    """
    if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
        raise TypeError("Both arguments must be numbers")
    if b == 0:
        raise ZeroDivisionError("Cannot divide by zero")
    return a / b
````

### Exemple Complet : Suivre Toutes les Meilleures Pratiques

````python
def calculate_grade_average(scores):
    """
    Calculate the average of a list of scores.
    
    Args:
        scores: List of numeric scores
    
    Returns:
        float: Average score
    
    Raises:
        TypeError: If scores is not a list
        ValueError: If scores is empty or contains invalid values
    """
    # Validate input type
    if not isinstance(scores, list):
        raise TypeError(f"Expected list, got {type(scores).__name__}")
    
    # Check for empty list
    if len(scores) == 0:
        raise ValueError("Cannot calculate average of empty list")
    
    # Calculate average with error handling
    try:
        total = sum(scores)
        average = total / len(scores)
    except TypeError as e:
        raise ValueError(f"All scores must be numbers: {e}")
    
    return average

# Test with valid input
try:
    avg = calculate_grade_average([85, 90, 78, 92, 88])
    print(f"Average: {avg:.2f}")
except (TypeError, ValueError) as e:
    print(f"Error: {e}")

# Test with invalid inputs
test_cases = [
    [],                           # Empty list
    "not a list",                 # Wrong type
    [85, "ninety", 78],           # Invalid score
]

for test in test_cases:
    try:
        avg = calculate_grade_average(test)
        print(f"Average: {avg}")
    except (TypeError, ValueError) as e:
        print(f"Error: {e}")
````

````python
def get_integer_input(prompt, min_value=None, max_value=None):
    """
    Get integer input from user with validation.
    
    Args:
        prompt: Message to show user
        min_value: Minimum allowed value (optional)
        max_value: Maximum allowed value (optional)
    
    Returns:
        int: Valid integer from user
    """
    while True:
        try:
            user_input = input(prompt)
            number = int(user_input)
            
            # Validate range
            if min_value is not None and number < min_value:
                print(f"Error: Value must be at least {min_value}")
                continue
            
            if max_value is not None and number > max_value:
                print(f"Error: Value must be at most {max_value}")
                continue
            
            return number
            
        except ValueError:
            print(f"Error: '{user_input}' is not a valid integer")
            print("Please enter a whole number (e.g., 42)\n")
        except KeyboardInterrupt:
            print("\nOperation cancelled by user")
            return None

# Test it (commented out for notebook)
# age = get_integer_input("Enter your age (0-120): ", min_value=0, max_value=120)
# print(f"You entered: {age}")

print("Function defined - test with actual user input in a Python script!")
````

### Application 2 : Opérations de Fichiers Sécurisées

````python
import os
from datetime import datetime

def safe_read_file(filename, default=None):
    """
    Safely read a file with comprehensive error handling.
    
    Args:
        filename: Path to file
        default: Value to return if file can't be read
    
    Returns:
        str: File contents or default value
    """
    try:
        with open(filename, "r") as file:
            content = file.read()
            return content
    
    except FileNotFoundError:
        print(f"File not found: {filename}")
        print(f"Current directory: {os.getcwd()}")
        return default
    
    except PermissionError:
        print(f"Permission denied: {filename}")
        print("You don't have permission to read this file")
        return default
    
    except UnicodeDecodeError:
        print(f"Cannot decode: {filename}")
        print("File may be binary or have invalid encoding")
        return default
    
    except Exception as e:
        print(f"Unexpected error reading {filename}: {e}")
        return default

def safe_write_file(filename, content, mode="w"):
    """
    Safely write to a file with error handling and backup.
    
    Args:
        filename: Path to file
        content: Content to write
        mode: Write mode ("w" or "a")
    
    Returns:
        bool: True if successful, False otherwise
    """
    # Create backup if file exists
    if os.path.exists(filename) and mode == "w":
        backup_name = f"{filename}.backup"
        try:
            import shutil
            shutil.copy2(filename, backup_name)
            print(f"Backup created: {backup_name}")
        except Exception as e:
            print(f"Warning: Could not create backup: {e}")
    
    # Write file
    try:
        with open(filename, mode) as file:
            file.write(content)
        print(f"Successfully wrote to {filename}")
        return True
    
    except PermissionError:
        print(f"Permission denied: Cannot write to {filename}")
        return False
    
    except OSError as e:
        print(f"OS error writing {filename}: {e}")
        return False
    
    except Exception as e:
        print(f"Unexpected error writing {filename}: {e}")
        return False

# Test the functions
content = safe_read_file("test.txt", default="Default content")
print(f"Content: {content}\n")

success = safe_write_file("output.txt", "Hello, World!\nThis is a test.")
print(f"Write successful: {success}")
````

### Application 3 : Calculatrice avec Gestion Complète des Erreurs

````python
class Calculator:
    """A robust calculator with error handling."""
    
    def add(self, a, b):
        """Add two numbers."""
        try:
            return float(a) + float(b)
        except (ValueError, TypeError) as e:
            raise ValueError(f"Cannot add {a} and {b}: {e}")
    
    def subtract(self, a, b):
        """Subtract b from a."""
        try:
            return float(a) - float(b)
        except (ValueError, TypeError) as e:
            raise ValueError(f"Cannot subtract {b} from {a}: {e}")
    
    def multiply(self, a, b):
        """Multiply two numbers."""
        try:
            return float(a) * float(b)
        except (ValueError, TypeError) as e:
            raise ValueError(f"Cannot multiply {a} and {b}: {e}")
    
    def divide(self, a, b):
        """Divide a by b."""
        try:
            a, b = float(a), float(b)
            
            if b == 0:
                raise ZeroDivisionError("Cannot divide by zero")
            
            return a / b
        
        except ValueError as e:
            raise ValueError(f"Invalid numbers for division: {e}")
        except ZeroDivisionError as e:
            raise ZeroDivisionError(f"Division error: {e}")
    
    def power(self, base, exponent):
        """Calculate base raised to exponent."""
        try:
            result = float(base) ** float(exponent)
            
            if result == float('inf'):
                raise OverflowError("Result is too large")
            
            return result
        
        except ValueError as e:
            raise ValueError(f"Invalid numbers for power operation: {e}")
        except OverflowError as e:
            raise OverflowError(f"Power operation overflow: {e}")
    
    def safe_calculate(self, operation, a, b):
        """Perform calculation with error handling."""
        operations = {
            "+": self.add,
            "-": self.subtract,
            "*": self.multiply,
            "/": self.divide,
            "**": self.power
        }
        
        try:
            if operation not in operations:
                raise ValueError(f"Unknown operation: {operation}")
            
            func = operations[operation]
            result = func(a, b)
            return result
        
        except (ValueError, ZeroDivisionError, OverflowError) as e:
            print(f"Error: {e}")
            return None

# Test the calculator
calc = Calculator()

test_cases = [
    ("+", 10, 5),      # Valid
    ("-", 10, 3),      # Valid
    ("*", 4, 5),       # Valid
    ("/", 20, 4),      # Valid
    ("/", 10, 0),      # Division by zero
    ("+", "10", "5"),  # String numbers (should work)
    ("*", "hello", 5),  # Invalid
    ("^", 2, 3),       # Unknown operation
]

print("Calculator Tests:\n")
for op, a, b in test_cases:
    result = calc.safe_calculate(op, a, b)
    if result is not None:
        print(f"{a} {op} {b} = {result}")
    print()
````

### Application 4 : Validateur de Données

````python
class DataValidator:
    """Validate various types of data with comprehensive error handling."""
    
    @staticmethod
    def validate_email(email):
        """Validate email format."""
        if not isinstance(email, str):
            raise TypeError("Email must be a string")
        
        email = email.strip().lower()
        
        if len(email) == 0:
            raise ValueError("Email cannot be empty")
        
        if "@" not in email:
            raise ValueError("Email must contain @")
        
        if email.count("@") != 1:
            raise ValueError("Email must contain exactly one @")
        
        username, domain = email.split("@")
        
        if len(username) == 0:
            raise ValueError("Email username part cannot be empty")
        
        if "." not in domain:
            raise ValueError("Email domain must contain a dot")
        
        return email
    
    @staticmethod
    def validate_age(age):
        """Validate age."""
        try:
            age = int(age)
        except (ValueError, TypeError):
            raise ValueError(f"Age must be a number, got {type(age).__name__}")
        
        if age < 0:
            raise ValueError(f"Age cannot be negative: {age}")
        
        if age > 150:
            raise ValueError(f"Age seems unrealistic: {age}")
        
        return age
    
    @staticmethod
    def validate_phone(phone):
        """Validate phone number (basic)."""
        if not isinstance(phone, str):
            raise TypeError("Phone must be a string")
        
        # Remove common separators
        digits = "".join(c for c in phone if c.isdigit())
        
        if len(digits) < 10:
            raise ValueError(f"Phone number too short: {phone}")
        
        if len(digits) > 15:
            raise ValueError(f"Phone number too long: {phone}")
        
        return phone
    
    @staticmethod
    def validate_user_data(data):
        """Validate complete user data."""
        if not isinstance(data, dict):
            raise TypeError("User data must be a dictionary")
        
        errors = []
        validated = {}
        
        # Validate email
        try:
            validated["email"] = DataValidator.validate_email(data.get("email", ""))
        except (ValueError, TypeError) as e:
            errors.append(f"Email error: {e}")
        
        # Validate age
        try:
            validated["age"] = DataValidator.validate_age(data.get("age"))
        except (ValueError, TypeError) as e:
            errors.append(f"Age error: {e}")
        
        # Validate phone
        try:
            validated["phone"] = DataValidator.validate_phone(data.get("phone", ""))
        except (ValueError, TypeError) as e:
            errors.append(f"Phone error: {e}")
        
        if errors:
            raise ValueError(f"Validation failed:\n" + "\n".join(errors))
        
        return validated

# Test the validator
validator = DataValidator()

# Valid data
valid_user = {
    "email": "alice@example.com",
    "age": 25,
    "phone": "555-123-4567"
}

try:
    result = validator.validate_user_data(valid_user)
    print("Valid user data:")
    for key, value in result.items():
        print(f"  {key}: {value}")
except ValueError as e:
    print(f"Validation error:\n{e}")

print("\n" + "="*50 + "\n")

# Invalid data
invalid_user = {
    "email": "invalid-email",
    "age": -5,
    "phone": "123"
}

try:
    result = validator.validate_user_data(invalid_user)
    print("This shouldn't print")
except ValueError as e:
    print("Invalid user data:")
    print(e)
````

### Référence Rapide - Try-Except :

**Basique :**
```python

```

**Exceptions Multiples :**
```python
try:
    operation()
except ValueError:
    gerer_erreur_valeur()
except TypeError:
    gerer_erreur_type()
```

**Avec else :**
```python
try:
    operation()
except ValueError:
    gerer_erreur()
else:
    # S'exécute si pas d'erreur
    action_succes()
```

**Avec finally :**
```python
try:
    operation()
except ValueError:
    gerer_erreur()
finally:
    # S'exécute toujours
    nettoyage()
```

**Lever des Exceptions :**
```python
def valider(valeur):
```

### Bonus : Exceptions Personnalisées (Optionnel/Avancé)

Pour les utilisateurs avancés, vous pouvez créer des classes d'exception personnalisées :

````python
# BONUS: Custom exceptions (optional - advanced topic)
class InsufficientFundsError(Exception):
    """Raised when account doesn't have enough money."""
    pass

class InvalidAccountError(Exception):
    """Raised when account number is invalid."""
    pass

class BankAccount:
    def __init__(self, account_number, balance=0):
        if not isinstance(account_number, str) or len(account_number) < 5:
            raise InvalidAccountError("Account number must be string with 5+ characters")
        self.account_number = account_number
        self.balance = balance
    
    def withdraw(self, amount):
        if amount > self.balance:
            raise InsufficientFundsError(
                f"Cannot withdraw ${amount}, balance is ${self.balance}"
            )
        self.balance -= amount
        return self.balance

# Using custom exceptions
try:
    account = BankAccount("ACC001", balance=100)
    account.withdraw(150)  # Not enough money
except InsufficientFundsError as e:
    print(f"Transaction failed: {e}")

try:
    bad_account = BankAccount("123")  # Too short
except InvalidAccountError as e:
    print(f"Account creation failed: {e}")

print("\nCustom exceptions let you create domain-specific error types!")
````
