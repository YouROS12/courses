## Partie 1 : Comprendre les Dictionnaires

Un **dictionnaire** stocke des données sous forme de **paires clé-valeur**. Pensez-y comme un vrai dictionnaire où vous cherchez un mot (clé) pour trouver sa définition (valeur).

```python
# Real-world analogy: A contact book
# Name (key) → Phone Number (value)

contacts = {
    "Alice": "555-1234",
    "Bob": "555-5678",
    "Carol": "555-9012"
}

print(contacts)
```

```python
# Why use dictionaries instead of lists?
# Lists use numeric indices (0, 1, 2...)
# Dictionaries use meaningful keys ("name", "age", "email"...)

# List approach - hard to remember what each position means
student_list = ["Ahmed", 20, "Cairo", "CS"]
print("Name:", student_list[0])  # Which index was name again?

# Dictionary approach - clear and readable!
student_dict = {
    "name": "Ahmed",
    "age": 20,
    "city": "Cairo",
    "major": "CS"
}
print("Name:", student_dict["name"])  # Much clearer!
```

## Partie 3 : Accéder et Modifier les Valeurs

Utilisez les clés pour accéder et changer les valeurs dans les dictionnaires.

### Accéder aux Valeurs

```python
# Using square brackets []
student = {
    "name": "Maria",
    "age": 22,
    "major": "Biology"
}

print("Name:", student["name"])
print("Age:", student["age"])
print("Major:", student["major"])
```

```python
# What happens if a key doesn't exist?
# Uncomment the line below to see the error:
# print(student["email"])  # KeyError: 'email'
```

```python
# Using get() method - SAFER!
student = {"name": "Maria", "age": 22, "major": "Biology"}

# If key exists, returns the value
print("Name:", student.get("name"))

# If key doesn't exist, returns None (no error!)
print("Email:", student.get("email"))

# You can specify a default value
print("Email:", student.get("email", "Not provided"))
print("Phone:", student.get("phone", "No phone number"))
```

### Modifier les Valeurs

```python
# Change an existing value
person = {"name": "John", "age": 25, "city": "Boston"}
print("Before:", person)

person["age"] = 26  # Update age
person["city"] = "New York"  # Update city
print("After:", person)
```

### Ajouter de Nouvelles Paires Clé-Valeur

```python
# Add new keys by assigning values
person = {"name": "Emma", "age": 28}
print("Before:", person)

person["city"] = "London"  # Add new key
person["job"] = "Designer"  # Add another key
print("After:", person)
```

### Retirer des Éléments

```python
# Using del keyword
person = {"name": "Tom", "age": 30, "city": "Paris", "job": "Teacher"}
print("Before:", person)

del person["job"]  # Remove the 'job' key
print("After del:", person)
```

```python
# Using pop() - removes and returns the value
person = {"name": "Tom", "age": 30, "city": "Paris"}
print("Before:", person)

city = person.pop("city")  # Remove and get the value
print(f"Removed city: {city}")
print("After pop:", person)
```

```python
# Using clear() - removes all items
person = {"name": "Tom", "age": 30, "city": "Paris"}
print("Before:", person)

person.clear()
print("After clear:", person)
```

## Partie 4 : Méthodes de Dictionnaire

Les dictionnaires ont de nombreuses méthodes intégrées utiles.

### keys() - Obtenir toutes les clés

```python
student = {"name": "Alice", "age": 20, "major": "CS", "gpa": 3.8}

# Get all keys
keys = student.keys()
print("Keys:", keys)
print("Type:", type(keys))

# Convert to list if needed
keys_list = list(student.keys())
print("Keys as list:", keys_list)
```

### values() - Obtenir toutes les valeurs

```python
student = {"name": "Alice", "age": 20, "major": "CS", "gpa": 3.8}

# Get all values
values = student.values()
print("Values:", values)

# Convert to list
values_list = list(student.values())
print("Values as list:", values_list)
```

### items() - Obtenir les paires clé-valeur

```python
student = {"name": "Alice", "age": 20, "major": "CS", "gpa": 3.8}

# Get all key-value pairs as tuples
items = student.items()
print("Items:", items)

# Convert to list of tuples
items_list = list(student.items())
print("Items as list:", items_list)
```

### update() - Ajouter plusieurs éléments à la fois

```python
student = {"name": "Bob", "age": 21}
print("Before:", student)

# Add multiple key-value pairs
student.update({"major": "Physics", "gpa": 3.5, "city": "Boston"})
print("After update:", student)

# update() also modifies existing keys
student.update({"age": 22, "gpa": 3.7})
print("After second update:", student)
```

### Vérifier si une Clé Existe

```python
# Using 'in' operator
student = {"name": "Carol", "age": 23, "major": "Math"}

print("Is 'name' in dictionary?", "name" in student)
print("Is 'email' in dictionary?", "email" in student)
print("Is 'age' NOT in dictionary?", "age" not in student)

# Use in conditional statements
if "email" in student:
    print("Email:", student["email"])
else:
    print("Email not found")
```

### Obtenir la Longueur du Dictionnaire

```python
# len() returns the number of key-value pairs
student = {"name": "David", "age": 24, "major": "Chemistry", "gpa": 3.6}

print("Number of items:", len(student))

empty = {}
print("Empty dictionary length:", len(empty))
```

## Partie 5 : Itérer à Travers les Dictionnaires

Il existe plusieurs façons de boucler à travers les dictionnaires.

### Boucler à Travers les Clés

```python
# Default: iterating through a dictionary loops through keys
student = {"name": "Emma", "age": 21, "major": "Art"}

print("Keys:")
for key in student:
    print(f"  {key}")
```

```python
# Explicitly using .keys()
student = {"name": "Emma", "age": 21, "major": "Art"}

print("Keys (explicit):")
for key in student.keys():
    print(f"  {key}")
```

### Boucler à Travers les Valeurs

```python
# Using .values()
student = {"name": "Frank", "age": 22, "major": "Music"}

print("Values:")
for value in student.values():
    print(f"  {value}")
```

### Boucler à Travers les Paires Clé-Valeur (Le Plus Courant !)

```python
# Using .items() - best way to access both key and value!
student = {"name": "Grace", "age": 20, "major": "History", "gpa": 3.9}

print("Key-Value Pairs:")
for key, value in student.items():
    print(f"  {key}: {value}")
```

```python
# Practical example: Display formatted information
person = {
    "name": "Ahmed Hassan",
    "age": 28,
    "city": "Dubai",
    "occupation": "Data Scientist",
    "salary": 85000
}

print("=" * 40)
print("PERSON INFORMATION")
print("=" * 40)
for key, value in person.items():
    # Capitalize and format the key
    formatted_key = key.replace("_", " ").title()
    print(f"{formatted_key:15s}: {value}")
print("=" * 40)
```

## Partie 6 : Dictionnaire vs Liste

Comprendre quand utiliser des dictionnaires vs des listes est important !

```python
# Lists: Use for ordered collections of similar items
shopping_list = ["milk", "bread", "eggs", "butter"]
scores = [85, 90, 78, 92, 88]

print("Shopping list:", shopping_list)
print("First item:", shopping_list[0])
print("Third score:", scores[2])
```

```python
# Dictionaries: Use for labeled data with different attributes
student = {
    "name": "John",
    "age": 20,
    "major": "CS",
    "gpa": 3.7,
    "courses": ["Math", "Physics", "Programming"]
}

print("Student info:", student)
print("Name:", student["name"])  # Clear what we're accessing
print("Courses:", student["courses"])
```

```python
# Comparison: Storing multiple students

# Using lists (harder to read and maintain)
student1 = ["Alice", 20, "Biology", 3.8]
student2 = ["Bob", 21, "Chemistry", 3.5]
# What does index 2 represent? Hard to remember!

# Using dictionaries (clear and self-documenting)
student1 = {"name": "Alice", "age": 20, "major": "Biology", "gpa": 3.8}
student2 = {"name": "Bob", "age": 21, "major": "Chemistry", "gpa": 3.5}
# Much clearer!

print(f"{student1['name']} is {student1['age']} years old")
```

### Quand Utiliser Quoi ?

**Utilisez des Listes quand :**
- Vous avez une collection d'éléments similaires
- L'ordre compte
- Vous devez accéder aux éléments par position
- Exemples : articles d'achat, scores de test, températures

**Utilisez des Dictionnaires quand :**
- Vous avez des données étiquetées/nommées
- Vous devez rechercher des valeurs par des clés significatives
- Vous stockez des attributs d'un objet
- Exemples : informations de personne, détails de produit, paramètres

## Partie 7 : Dictionnaires Imbriqués

Les dictionnaires peuvent contenir d'autres dictionnaires (ou listes) !

```python
# Simple nested dictionary
student = {
    "name": "Lisa",
    "age": 21,
    "grades": {
        "math": 90,
        "english": 85,
        "science": 92
    }
}

# Access nested values
print("Student name:", student["name"])
print("Math grade:", student["grades"]["math"])
print("English grade:", student["grades"]["english"])
```

```python
# Multiple students with nested information
students = {
    "student1": {
        "name": "Alice",
        "age": 20,
        "major": "CS"
    },
    "student2": {
        "name": "Bob",
        "age": 21,
        "major": "Math"
    },
    "student3": {
        "name": "Carol",
        "age": 22,
        "major": "Physics"
    }
}

# Access specific student
print("Student 1 name:", students["student1"]["name"])
print("Student 2 major:", students["student2"]["major"])

# Loop through all students
print("\nAll Students:")
for student_id, info in students.items():
    print(f"\n{student_id}:")
    for key, value in info.items():
        print(f"  {key}: {value}")
```

```python
# Dictionary with lists
company = {
    "name": "Tech Solutions",
    "employees": ["Alice", "Bob", "Carol", "David"],
    "departments": ["Engineering", "Sales", "HR"],
    "location": {
        "city": "San Francisco",
        "state": "CA",
        "zip": "94102"
    }
}

print("Company:", company["name"])
print("First employee:", company["employees"][0])
print("City:", company["location"]["city"])
print("All employees:", company["employees"])
```

Construisons des programmes pratiques utilisant des dictionnaires !

### Application 1 : Annuaire Téléphonique

```python
# Simple phone book application
phone_book = {
    "Alice": "555-1234",
    "Bob": "555-5678",
    "Carol": "555-9012",
    "David": "555-3456"
}

# Look up a contact
name = input("Enter name to look up: ")

if name in phone_book:
    print(f"{name}'s phone number: {phone_book[name]}")
else:
    print(f"{name} not found in phone book")

# Display all contacts
print("\n" + "=" * 30)
print("ALL CONTACTS")
print("=" * 30)
for name, phone in phone_book.items():
    print(f"{name:15s}: {phone}")
```

### Application 2 : Suivi de Notes d'Étudiant

```python
# Student with multiple subject grades
student = {
    "name": "Ahmed",
    "student_id": "S12345",
    "grades": {
        "Mathematics": 88,
        "Physics": 92,
        "Chemistry": 85,
        "English": 90,
        "Computer Science": 95
    }
}

# Display student report
print("=" * 50)
print(f"STUDENT GRADE REPORT")
print("=" * 50)
print(f"Name: {student['name']}")
print(f"Student ID: {student['student_id']}")
print("\nGrades:")
print("-" * 50)

total = 0
for subject, grade in student["grades"].items():
    print(f"{subject:20s}: {grade}")
    total += grade

average = total / len(student["grades"])
print("-" * 50)
print(f"Average Grade: {average:.2f}")
print("=" * 50)
```

### Application 3 : Système de Gestion d'Inventaire

```python
# Store inventory
inventory = {
    "Laptop": {"price": 999.99, "quantity": 15, "category": "Electronics"},
    "Mouse": {"price": 25.99, "quantity": 50, "category": "Electronics"},
    "Keyboard": {"price": 79.99, "quantity": 30, "category": "Electronics"},
    "Monitor": {"price": 299.99, "quantity": 20, "category": "Electronics"},
    "Headphones": {"price": 149.99, "quantity": 25, "category": "Audio"}
}

# Display inventory
print("=" * 70)
print("INVENTORY REPORT")
print("=" * 70)
print(f"{'Product':<15} {'Price':>10} {'Quantity':>10} {'Value':>12} {'Category':<15}")
print("-" * 70)

total_value = 0
for product, details in inventory.items():
    value = details["price"] * details["quantity"]
    total_value += value
    print(f"{product:<15} ${details['price']:>9.2f} {details['quantity']:>10} ${value:>11.2f} {details['category']:<15}")

print("=" * 70)
print(f"Total Inventory Value: ${total_value:,.2f}")
print(f"Total Products: {len(inventory)}")
print("=" * 70)
```

### Application 4 : Compteur de Mots

```python
# Count word frequency in a sentence
text = "python is amazing and python is powerful and python is fun"

# Split text into words
words = text.lower().split()

# Count each word
word_count = {}
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

# Display results
print("Word Frequency:")
for word, count in word_count.items():
    print(f"  {word}: {count}")

# Find most common word
most_common = max(word_count, key=word_count.get)
print(f"\nMost common word: '{most_common}' appears {word_count[most_common]} times")
```
