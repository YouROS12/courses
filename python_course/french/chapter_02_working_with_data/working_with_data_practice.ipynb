## Partie 3 : Conversion de Type - La Solution ! 🔄

Nous pouvons convertir des chaînes en nombres (et vice versa) en utilisant des fonctions de conversion de type.

### Conversion en Entier - int()

```python
# String to integer
text_number = "42"
real_number = int(text_number)

print("Original:", text_number, "Type:", type(text_number))
print("Converted:", real_number, "Type:", type(real_number))
```

```python
# Now we can do math!
text_num = "25"
number = int(text_num)

result = number + 10
print(f"{number} + 10 = {result}")
```

```python
# Float to int (removes decimal part)
decimal = 3.9
whole = int(decimal)

print(f"{decimal} becomes {whole}")  # 3.9 becomes 3
```

### Conversion en Décimal - float()

```python
# String to float
text_price = "19.99"
price = float(text_price)

print("Original:", text_price, "Type:", type(text_price))
print("Converted:", price, "Type:", type(price))
```

```python
# Integer to float
whole_number = 5
decimal_number = float(whole_number)

print(f"{whole_number} becomes {decimal_number}")  # 5 becomes 5.0
```

### Conversion en Chaîne - str()

```python
# Number to string
age = 25
age_text = str(age)

print("Original:", age, "Type:", type(age))
print("Converted:", age_text, "Type:", type(age_text))
```

```python
# Why convert to string? For concatenation!
score = 100
message = "Your score is " + str(score) + " points!"
print(message)

# MODERN WAY: Use f-strings instead (much cleaner!)
score = 100
message = f"Your score is {score} points!"
print(message)

# f-strings are the recommended way to format strings in Python!
```

## Partie 5 : Opérations sur les Chaînes ✂️

Les chaînes ont de nombreuses méthodes et opérations utiles !

### Concaténation de Chaînes (Jonction)

```python
# Using + operator
first_name = "Ahmed"
last_name = "Hassan"

full_name = first_name + " " + last_name
print(full_name)
```

```python
# Building sentences
greeting = "Hello"
name = "Sarah"
punctuation = "!"

message = greeting + " " + name + punctuation
print(message)  # Hello Sarah!
```

### Répétition de Chaînes

```python
# Repeat strings with *
print("Ha" * 5)  # HaHaHaHaHa
print("=" * 20)  # ====================
print("Python! " * 3)  # Python! Python! Python!
```

```python
# Create visual separators
print("*" * 30)
print("      WELCOME TO PYTHON")
print("*" * 30)
```

### Méthodes de Chaînes

```python
# upper() - Convert to uppercase
text = "hello world"
print(text.upper())  # HELLO WORLD
```

```python
# lower() - Convert to lowercase
text = "HELLO WORLD"
print(text.lower())  # hello world
```

```python
# title() - Capitalize first letter of each word
text = "python programming is fun"
print(text.title())  # Python Programming Is Fun
```

```python
# capitalize() - Capitalize only first letter
text = "python programming"
print(text.capitalize())  # Python programming
```

```python
# strip() - Remove leading/trailing spaces
text = "   hello   "
print(f"'{text}'")  # '   hello   '
print(f"'{text.strip()}'")  # 'hello'
```

```python
# replace() - Replace parts of a string
text = "I love cats"
new_text = text.replace("cats", "dogs")
print(new_text)  # I love dogs
```

```python
# count() - Count occurrences
text = "hello hello hello"
count = text.count("hello")
print(f"'hello' appears {count} times")
```

```python
# len() - Get string length
text = "Python"
length = len(text)
print(f"'{text}' has {length} characters")
```

## Partie 7 : Comprendre les Erreurs 🐛

Les erreurs sont vos amies - elles vous disent ce qui ne va pas !

### ValueError - Conversion Invalide

```python
# This will cause a ValueError
# Uncomment to see the error:
# number = int("hello")  # Can't convert "hello" to integer!
```

```python
# This works fine
number = int("123")  # "123" can be converted to 123
print(number)
```

### TypeError - Types Incompatibles

```python
# This will cause a TypeError
# Uncomment to see the error:
# result = "5" + 5  # Can't add string and integer!
```

```python
# Fix it by converting
result1 = int("5") + 5  # Convert string to int
print(result1)  # 10

result2 = "5" + str(5)  # Convert int to string
print(result2)  # "55"
```

### SyntaxError - Code Python Invalide

```python
# This will cause a SyntaxError
# Uncomment to see the error:
# print("Hello"  # Missing closing parenthesis
```

```python
# Fixed version
print("Hello")  # All parentheses match!
```

## Partie 8 : Applications Réelles 🌍

Construisons des programmes interactifs !

### Exemple 1 : Salutation Personnalisée

```python
# Get user information
first_name = input("Enter your first name: ")
last_name = input("Enter your last name: ")

# Format nicely
full_name = (first_name + " " + last_name).title()

# Create personalized greeting
print("\n" + "=" * 40)
print(f"  Welcome, {full_name}!")
print("=" * 40)
```

### Exemple 2 : Âge en Différentes Unités

```python
# Get age from user
age_years = int(input("How old are you? "))

# Convert to different units
age_months = age_years * 12
age_days = age_years * 365  # Approximate
age_hours = age_days * 24

# Display results
print(f"\nYou are:")
print(f"  {age_years} years old")
print(f"  {age_months} months old")
print(f"  {age_days} days old")
print(f"  {age_hours:,} hours old!")  # :, adds comma formatting
```

### Exemple 3 : Convertisseur de Température

```python
# Celsius to Fahrenheit converter
celsius = float(input("Enter temperature in Celsius: "))

# Convert: F = (C × 9/5) + 32
fahrenheit = (celsius * 9/5) + 32

# Display result
print(f"\n{celsius}°C = {fahrenheit}°F")
```

### Exemple 4 : Diviseur de Note de Restaurant

```python
# Get bill information
total_bill = float(input("Enter total bill amount: $"))
num_people = int(input("How many people? "))
tip_percent = float(input("Tip percentage (e.g., 15): "))

# Calculate
tip_amount = total_bill * (tip_percent / 100)
total_with_tip = total_bill + tip_amount
per_person = total_with_tip / num_people

# Display results
print("\n" + "=" * 40)
print(f"Original Bill: ${total_bill:.2f}")
print(f"Tip ({tip_percent}%): ${tip_amount:.2f}")
print(f"Total: ${total_with_tip:.2f}")
print(f"Per Person: ${per_person:.2f}")
print("=" * 40)
```

### Exemple 5 : Générateur de Badge Nominatif

```python
# Get user info
name = input("Enter your name: ")
title = input("Enter your title/role: ")
company = input("Enter company name: ")

# Format nicely
name_upper = name.upper()
title_title = title.title()
company_caps = company.upper()

# Create name tag
print("\n" + "*" * 50)
print("*" + " " * 48 + "*")
print(f"*  HELLO, MY NAME IS:")
print(f"*  {name_upper}")
print("*")
print(f"*  {title_title}")
print(f"*  {company_caps}")
print("*" + " " * 48 + "*")
print("*" * 50)
```
