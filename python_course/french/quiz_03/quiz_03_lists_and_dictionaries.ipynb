### Défi 2 : Statistiques de liste - Exemple de solution
```python
# Créer la liste de scores
scores = [85, 92, 78, 90, 88]

# Afficher le score le plus élevé
print("Score le plus élevé:", max(scores))

# Afficher le score le plus bas
print("Score le plus bas:", min(scores))

# Afficher le total de tous les scores
print("Total:", sum(scores))

# Calculer et afficher la moyenne
average = sum(scores) / len(scores)
print("Moyenne:", average)

# Vérifier si 90 est dans la liste
print("90 est-il dans la liste ?", 90 in scores)
```
