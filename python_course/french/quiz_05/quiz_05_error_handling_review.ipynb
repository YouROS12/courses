### Défi 3 : Notes d'étudiants - Exemple de solution
```python
# Créer le fichier de notes
with open("grades.txt", "w") as file:
    file.write("Alice,85\n")
    file.write("Bob,92\n")
    file.write("Charlie,78\n")
    file.write("Diana,95\n")

# Lire et traiter le fichier
students = []
grades = []

with open("grades.txt", "r") as file:
    for line in file:
        line = line.strip()  # Supprimer le caractère de nouvelle ligne
        name, grade = line.split(",")
        students.append(name)
        grades.append(int(grade))

# Calculer les statistiques
average = sum(grades) / len(grades)
highest = max(grades)
lowest = min(grades)

# Trouver les étudiants avec les notes les plus élevées et les plus basses
highest_student = students[grades.index(highest)]
lowest_student = students[grades.index(lowest)]

# Afficher les résultats
print(f"Note moyenne : {average:.2f}")
print(f"Note la plus élevée : {highest} ({highest_student})")
print(f"Note la plus basse : {lowest} ({lowest_student})")

# Créer le fichier de rapport de notes
with open("grade_report.txt", "w") as file:
    file.write("Rapport de notes\n")
    file.write("=" * 40 + "\n")
    file.write(f"Note moyenne : {average:.2f}\n")
    file.write(f"Note la plus élevée : {highest} ({highest_student})\n")
    file.write(f"Note la plus basse : {lowest} ({lowest_student})\n")
    file.write("\nTous les étudiants :\n")
    for i in range(len(students)):
        file.write(f"{students[i]}: {grades[i]}\n")

print("\nRapport de notes enregistré dans 'grade_report.txt'")
```
