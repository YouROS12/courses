### Challenge 4: Student Grades - Sample Solution
```python
# Create grades dictionary
grades = {"Alice": 92, "Bob": 87, "Charlie": 95, "Diana": 88}

# Calculate average
average = sum(grades.values()) / len(grades)
print("Average score:", average)

# Find highest score
highest_score = max(grades.values())
for student, score in grades.items():
    if score == highest_score:
        print(f"Highest: {student} with {score}")
        break

# Find lowest score
lowest_score = min(grades.values())
for student, score in grades.items():
    if score == lowest_score:
        print(f"Lowest: {student} with {score}")
        break

# Print all students with scores
print("\nAll students:")
for student, score in grades.items():
    print(f"{student}: {score}")
```
