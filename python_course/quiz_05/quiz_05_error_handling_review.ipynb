### Challenge 3: Student Grades - Sample Solution
```python
# Create the grades file
with open("grades.txt", "w") as file:
    file.write("Alice,85\n")
    file.write("Bob,92\n")
    file.write("Charlie,78\n")
    file.write("Diana,95\n")

# Read and process the file
students = []
grades = []

with open("grades.txt", "r") as file:
    for line in file:
        line = line.strip()  # Remove newline character
        name, grade = line.split(",")
        students.append(name)
        grades.append(int(grade))

# Calculate statistics
average = sum(grades) / len(grades)
highest = max(grades)
lowest = min(grades)

# Find students with highest and lowest grades
highest_student = students[grades.index(highest)]
lowest_student = students[grades.index(lowest)]

# Print results
print(f"Average grade: {average:.2f}")
print(f"Highest grade: {highest} ({highest_student})")
print(f"Lowest grade: {lowest} ({lowest_student})")

# Create grade report file
with open("grade_report.txt", "w") as file:
    file.write("Grade Report\n")
    file.write("=" * 40 + "\n")
    file.write(f"Average grade: {average:.2f}\n")
    file.write(f"Highest grade: {highest} ({highest_student})\n")
    file.write(f"Lowest grade: {lowest} ({lowest_student})\n")
    file.write("\nAll Students:\n")
    for i in range(len(students)):
        file.write(f"{students[i]}: {grades[i]}\n")

print("\nGrade report saved to 'grade_report.txt'")
```

**Debug 1:** Tuples are immutable - you cannot change their values

**Debug 2:** Sets don't support indexing - they are unordered
```python
numbers = {1, 2, 3, 4, 5}
# Solution 1: Convert to list first
numbers_list = list(numbers)
print(numbers_list[2])

# Solution 2: Check membership instead
print(3 in numbers)

# Solution 3: Iterate through the set
for num in numbers:
    print(num)
```

**Debug 3:** Sets use `add()` method, not `append()`
```python
my_set = {1, 2, 3}
my_set.add(4)  # Use add() instead of append()
print(my_set)
```

**Debug 4:** File was not closed - always close files after use
