## Part 4: Basic Math Operations ➕➖✖️➗

Python can be used as a powerful calculator!

```python
# Addition
print(10 + 5)        # 15
print(100 + 250)     # 350
```

```python
# Subtraction
print(20 - 8)        # 12
print(100 - 45)      # 55
```

```python
# Multiplication
print(6 * 7)         # 42
print(12 * 5)        # 60
```

```python
# Division (always gives a float)
print(20 / 4)        # 5.0
print(15 / 2)        # 7.5
```

```python
# Power (exponentiation)
print(2 ** 3)        # 2 × 2 × 2 = 8
print(5 ** 2)        # 5 × 5 = 25
print(10 ** 3)       # 10 × 10 × 10 = 1000
```

```python
# Modulo (remainder)
print(10 % 3)        # 1 (because 10 ÷ 3 = 3 remainder 1)
print(17 % 5)        # 2 (because 17 ÷ 5 = 3 remainder 2)
print(20 % 4)        # 0 (because 20 ÷ 4 = 5 remainder 0)
```

```python
# Integer division (rounds down)
print(17 // 5)       # 3 (ignores the remainder)
print(20 // 3)       # 6
```

## Part 5: Operations with Variables 🧮

```python
# Using variables in calculations
length = 10
width = 5

area = length * width
perimeter = 2 * (length + width)

print("Rectangle dimensions:", length, "x", width)
print("Area:", area)
print("Perimeter:", perimeter)
```

```python
# Updating variables using operations
score = 10
print("Initial score:", score)

score = score + 5      # Add 5 to score
print("After bonus:", score)

score = score * 2      # Double the score
print("After doubling:", score)
```

```python
# Shorthand operators
count = 10
print("Initial:", count)

count += 5    # Same as: count = count + 5
print("After += 5:", count)

count -= 3    # Same as: count = count - 3
print("After -= 3:", count)

count *= 2    # Same as: count = count * 2
print("After *= 2:", count)
```

## Part 6: Working with Strings 📝

```python
# Creating strings
message1 = "Hello, Python!"
message2 = 'Single quotes work too!'
message3 = "You can use 'single quotes' inside double quotes"

print(message1)
print(message2)
print(message3)
```

```python
# String concatenation (joining strings)
first_name = "Ahmed"
last_name = "Hassan"

full_name = first_name + " " + last_name
print(full_name)
```

```python
# Repeating strings
laugh = "ha" * 5
print(laugh)          # hahahahahaha

separator = "=" * 20
print(separator)
```

```python
# Combining text and numbers in print
name = "Fatima"
age = 22
height = 1.68

print("Name:", name)
print("Age:", age, "years old")
print("Height:", height, "meters")
```

```python
# String formatting with f-strings (modern way)
name = "Youssef"
age = 25
city = "Marrakech"

message = f"My name is {name}, I am {age} years old, and I live in {city}"
print(message)
```

## Part 7: Real-World Applications 🌍

Let's apply what we've learned to solve real problems!

### Example 1: Shopping Receipt Calculator

```python
# Shopping items and prices
item1 = "Notebook"
price1 = 25.50

item2 = "Pen"
price2 = 5.00

item3 = "Backpack"
price3 = 150.00

# Calculate total
total = price1 + price2 + price3

# Display receipt
print("========== RECEIPT ==========")
print(f"{item1}: ${price1}")
print(f"{item2}: ${price2}")
print(f"{item3}: ${price3}")
print("=============================")
print(f"TOTAL: ${total}")
print("=============================")
```

### Example 2: Temperature Converter (Celsius to Fahrenheit)

```python
# Temperature in Celsius
celsius = 25

# Convert to Fahrenheit: F = (C × 9/5) + 32
fahrenheit = (celsius * 9/5) + 32

print(f"{celsius}°C is equal to {fahrenheit}°F")
```

### Example 3: Age Calculator

```python
# Calculate age in different units
age_years = 20

age_months = age_years * 12
age_days = age_years * 365  # Approximate
age_hours = age_days * 24

print(f"You are {age_years} years old")
print(f"That's {age_months} months")
print(f"Or {age_days} days")
print(f"Or approximately {age_hours} hours!")
```

### Example 4: Simple Tip Calculator

```python
# Restaurant bill
bill_amount = 250.00
tip_percentage = 15  # 15% tip

# Calculate tip and total
tip_amount = bill_amount * (tip_percentage / 100)
total_amount = bill_amount + tip_amount

print(f"Bill Amount: ${bill_amount}")
print(f"Tip ({tip_percentage}%): ${tip_amount}")
print(f"Total to Pay: ${total_amount}")
```
